```python
def download_json(filename):
    path = hf_hub_download(repo_id=DATA_REPO, filename=filename, repo_type='dataset', token=HF_TOKEN)
    with open(path, encoding='utf-8') as f: return json.load(f)

def normalize_documents(raw, labelled):
    if isinstance(raw, list): docs = raw
    elif isinstance(raw, dict) and 'id' in raw and 'document' in raw: docs = [raw]
    elif isinstance(raw, dict):
        docs = []
        for doc_id, value in raw.items():
            if not isinstance(value, dict): raise ValueError('Expected ID -> document object mapping')
            doc = dict(value)
            doc.setdefault('id', doc_id)
            if doc['id'] != doc_id: raise ValueError('Document ID mismatch')
            docs.append(doc)
    else: raise ValueError('Expected list of document objects')
    result, excluded = [], []
    seen_ids = set()
    for doc in docs:
        doc_id, text = doc['id'], doc['document']
        if not isinstance(doc_id, str) or not isinstance(text, str) or not text.strip():
            raise ValueError('Expected string ID and nonempty document text')
        if doc_id in seen_ids: raise ValueError(f'Duplicate document ID: {doc_id}')
        seen_ids.add(doc_id)
        row = dict(id=doc_id, document=text)
        if labelled:
            if SUMMARY_KEY not in doc:
                raise ValueError(f'{doc_id}: summary field {SUMMARY_KEY!r} is absent. Available fields: {list(doc)}. Set SUMMARY_KEY to the actual reference field; missing fields are not silently skipped.')
            target = doc[SUMMARY_KEY]
            if target is None or (isinstance(target, str) and not target.strip()):
                if not SKIP_EMPTY_SUMMARIES:
                    raise ValueError(f'{doc_id}: null or blank reference summary')
                excluded.append(dict(id=doc_id, reason='null' if target is None else 'blank'))
                continue
            if not isinstance(target, str):
                raise ValueError(f'{doc_id}: {SUMMARY_KEY!r} must contain a string; got {type(target).__name__}. Inspect this record rather than coercing the target.')
            row['summary'] = target
        result.append(row)
    if labelled:
        print(f'Training input records: {len(docs):,} | usable references: {len(result):,} | excluded null/blank summaries: {len(excluded):,}')
        if excluded: print('Excluded IDs (first 20):', [r['id'] for r in excluded[:20]])
        with open(WORK_DIR / 'excluded_training_summaries.json', 'w') as f:
            json.dump(excluded, f, indent=2)
    if not result: raise ValueError('No usable records remain; inspect the dataset reference field')
    return result

all_train = normalize_documents(download_json(f'{DATA_PREFIX}/Train/train.json'), labelled=True)
# Exact duplicate document texts stay together, even if IDs differ.
groups = {}
for row in all_train:
    digest = hashlib.sha256(' '.join(row['document'].split()).encode()).hexdigest()
    groups.setdefault(digest, []).append(row)
keys = sorted(groups)
random.Random(SEED).shuffle(keys)
if len(keys) < 2: raise ValueError('At least two distinct documents are required')
assert 0 < VAL_FRACTION < 1 and 0 < TRAIN_FRACTION <= 1
n_dev = min(len(keys)-1, max(1, round(len(keys)*VAL_FRACTION)))
dev_keys, train_keys = keys[:n_dev], keys[n_dev:]
train_keys = train_keys[:max(1, round(len(train_keys)*TRAIN_FRACTION))]
data = {'train': [d for k in train_keys for d in groups[k]],
        'dev': [d for k in dev_keys for d in groups[k]]}
# Stable fixed subset, independent of generated outputs.
if VAL_MAX_DOCS is not None: data['dev'] = data['dev'][:VAL_MAX_DOCS]
assert data['dev'] and not ({d['id'] for d in data['train']} & {d['id'] for d in data['dev']})
DATA_FINGERPRINT = hashlib.sha256(json.dumps(all_train, sort_keys=True, ensure_ascii=False).encode()).hexdigest()
with open(WORK_DIR / 'split_ids.json', 'w') as f:
    json.dump({s: [d['id'] for d in rows] for s, rows in data.items()}, f, indent=2)
print('Public training records:', len(all_train), '| train:', len(data['train']), '| validation used:', len(data['dev']))
del all_train, groups
```

```python
def split_document_tokens(tokens, chunk_size=MAX_SEQ_LEN):
    if chunk_size <= 0: raise ValueError('Chunk size must be positive')
    return [tokens[i:i+chunk_size] for i in range(0,len(tokens),chunk_size)] or [[EOS_ID]]

def neighbor_indices(count):
    # Missing boundary neighbors are omitted, not wrapped around the document.
    return [[j for j in (i-1,i+1) if 0 <= j < count] for i in range(count)]

def neighbor_context_tokens(chunks,index):
    previous = chunks[index-1][-CONTEXT_TOKENS_PER_SIDE:] if index>0 else []
    following = chunks[index+1][:CONTEXT_TOKENS_PER_SIDE] if index+1<len(chunks) else []
    # EOS is an existing separator token, not a target or added vocabulary item.
    return previous+([EOS_ID] if previous and following else [])+following

class SummaryDataset(Dataset):
    def __init__(self, documents, split):
        self.rows, self.split = [], split
        source_lengths, target_lengths, chunk_counts = [], [], []
        for doc in tqdm(documents, desc=f'Tokenizing {split}'):
            source = tokenizer.encode(doc['document'], add_special_tokens=False).ids
            chunks = split_document_tokens(source)
            row = dict(doc_id=doc['id'], chunks=chunks)
            source_lengths.append(len(source)); chunk_counts.append(len(chunks))
            if 'summary' in doc:
                target = tokenizer.encode(doc['summary'], add_special_tokens=False).ids
                if not target: raise ValueError(f"Empty tokenized summary: {doc['id']}")
                target_lengths.append(len(target)+1)
                row.update(target=target[:MAX_NEW_TOKENS-1]+[EOS_ID], reference=doc['summary'])
            self.rows.append(row)
        for name, lengths in [('document',source_lengths),('summary including EOS',target_lengths)]:
            if lengths:
                print(split,name,'tokens p50/p90/p99/max:',torch.quantile(torch.tensor(lengths).float(),torch.tensor([.5,.9,.99,1.])).tolist())
        print(split,'documents/chunks:',len(self.rows),sum(chunk_counts),'| maximum chunks/document:',max(chunk_counts))
        print('All source chunks are retained; no source truncation or passage selection')
        if target_lengths:
            print(split,'summaries truncated for training/loss:',sum(x>MAX_NEW_TOKENS for x in target_lengths),'/',len(target_lengths))
    def __len__(self): return len(self.rows)
    def __getitem__(self,index): return self.rows[index]

def collate_summary(rows):
    if len(rows) != 1: raise ValueError('Use BATCH_SIZE=1: each backward covers one complete document')
    row = rows[0]
    target = row['target']
    decoder_ids = torch.tensor([[DECODER_START_ID]+target[:-1]],dtype=torch.long)
    return dict(chunks=row['chunks'], decoder_ids=decoder_ids,
                decoder_mask=torch.ones_like(decoder_ids,dtype=torch.bool),
                labels=torch.tensor([target],dtype=torch.long),doc_ids=[row['doc_id']])

datasets = {s:SummaryDataset(rows,s) for s,rows in data.items()}
def make_loader(split):
    return DataLoader(datasets[split],batch_size=BATCH_SIZE,shuffle=(split=='train'),
                      collate_fn=collate_summary,num_workers=0,pin_memory=(DEVICE=='cuda'))
```

```python
RMS_NORM_EPS = 1e-6
ROPE_BASE = 10000
DIM=640
ENCODER_LAYERS=6
DECODER_LAYERS=6
N_HEADS=16
N_KV_HEADS=4
FF_DIM=1728
MAX_ENCODER_LEN=512
MAX_DECODER_LEN=256
import torch
import torch.nn as nn
import torch.nn.functional as F

# ============================================================
# RMSNorm
# ============================================================

class RMSNorm(nn.Module):

    def __init__(self, dim, eps=RMS_NORM_EPS):
        super().__init__()

        self.weight = nn.Parameter(
            torch.ones(dim)
        )

        self.eps = eps

    def forward(self, x):

        rms = (x*x).mean(
            dim=-1,
            keepdim=True
        )

        return self.weight * x * torch.rsqrt(
            rms + self.eps
        )


# ============================================================
# RoPE
# ============================================================

class RotaryEmbedding(nn.Module):

    def __init__(
        self,
        head_dim,
        max_seq_len,
        base=ROPE_BASE,
    ):
        super().__init__()

        inv_freq = 1.0 / (
            base ** (
                torch.arange(
                    0,
                    head_dim,
                    2,
                    dtype=torch.float32,
                ) / head_dim
            )
        )

        positions = torch.arange(
            max_seq_len,
            dtype=torch.float32,
        )

        freqs = torch.outer(
            positions,
            inv_freq,
        )

        # [T, head_dim]
        emb = torch.cat(
            [freqs, freqs],
            dim=-1,
        )

        self.register_buffer(
            "cos",
            emb.cos()[None, None, :, :],
            persistent=False,
        )

        self.register_buffer(
            "sin",
            emb.sin()[None, None, :, :],
            persistent=False,
        )

    def forward(self, seq_len):

        return (
            self.cos[:, :, :seq_len],
            self.sin[:, :, :seq_len],
        )


def rotate_half(x):

    x1 = x[..., :x.shape[-1] // 2]
    x2 = x[..., x.shape[-1] // 2:]

    return torch.cat(
        [-x2, x1],
        dim=-1,
    )


def apply_rope(q, k, cos, sin):

    q = (
        q * cos
        +
        rotate_half(q) * sin
    )

    k = (
        k * cos
        +
        rotate_half(k) * sin
    )

    return q, k


# ============================================================
# SwiGLU
# ============================================================

class SwiGLU(nn.Module):

    def __init__(
        self,
        dim,
        hidden_dim,
    ):
        super().__init__()

        self.gate_proj = nn.Linear(
            dim,
            hidden_dim,
            bias=False,
        )

        self.up_proj = nn.Linear(
            dim,
            hidden_dim,
            bias=False,
        )

        self.down_proj = nn.Linear(
            hidden_dim,
            dim,
            bias=False,
        )

    def forward(self, x):

        return self.down_proj(
            F.silu(
                self.gate_proj(x)
            )
            *
            self.up_proj(x)
        )


# ============================================================
# GQA + Flash Attention
# ============================================================

class Attention(nn.Module):

    def __init__(
        self,
        dim,
        n_heads,
        n_kv_heads,
        max_seq_len,
        causal=False,
        use_rope=True,
    ):
        super().__init__()

        assert dim % n_heads == 0
        assert n_heads % n_kv_heads == 0

        self.dim = dim
        self.n_heads = n_heads
        self.n_kv_heads = n_kv_heads

        self.head_dim = dim // n_heads

        self.causal = causal
        self.use_rope = use_rope

        # ----------------------------------------------------
        # Q / K / V projections
        # ----------------------------------------------------

        self.q_proj = nn.Linear(
            dim,
            n_heads * self.head_dim,
            bias=False,
        )

        self.k_proj = nn.Linear(
            dim,
            n_kv_heads * self.head_dim,
            bias=False,
        )

        self.v_proj = nn.Linear(
            dim,
            n_kv_heads * self.head_dim,
            bias=False,
        )

        self.o_proj = nn.Linear(
            n_heads * self.head_dim,
            dim,
            bias=False,
        )

        # RoPE only needs to be constructed for self-attention
        if use_rope:

            self.rope = RotaryEmbedding(
                self.head_dim,
                max_seq_len,
            )

    def forward(
        self,
        x,
        kv=None,
        past_kv=None,
        use_cache=False,
        key_mask=None,
    ):

        B, T, C = x.shape

        # ====================================================
        # Queries
        # ====================================================

        q = self.q_proj(x)

        q = q.view(
            B,
            T,
            self.n_heads,
            self.head_dim,
        )

        q = q.transpose(1, 2)

        # ====================================================
        # Keys / Values
        # ====================================================

        # Self-attention:
        #     source = x
        #
        # Cross-attention:
        #     source = encoder output

        source = x if kv is None else kv

        S = source.shape[1]

        k = self.k_proj(source)
        v = self.v_proj(source)

        k = k.view(
            B,
            S,
            self.n_kv_heads,
            self.head_dim,
        )

        v = v.view(
            B,
            S,
            self.n_kv_heads,
            self.head_dim,
        )

        k = k.transpose(1, 2)
        v = v.transpose(1, 2)

        # ====================================================
        # RoPE
        # ====================================================

        # Only self-attention gets RoPE.
        # Cross-attention does NOT use RoPE.

        if kv is None and self.use_rope:

            current_len = k.shape[2]
            offset = 0 if past_kv is None else past_kv[0].shape[2]
            if offset + current_len > self.rope.cos.size(2):
                raise ValueError('Self-attention exceeds the Phase 1 positional context')
            cos, sin = self.rope(offset + current_len)
            cos = cos[:, :, offset:offset+current_len]
            sin = sin[:, :, offset:offset+current_len]

            q, k = apply_rope(
                q,
                k,
                cos,
                sin,
            )

        # ====================================================
        # Decoder KV Cache
        # ====================================================

        if (
            kv is None
            and past_kv is not None
        ):

            past_k, past_v = past_kv

            k = torch.cat(
                [past_k, k],
                dim=2,
            )

            v = torch.cat(
                [past_v, v],
                dim=2,
            )

        # Store native KV heads before expanding them for GQA.
        cache_kv = (k, v) if use_cache and kv is None else None

        # ====================================================
        # GQA
        # ====================================================

        if self.n_kv_heads != self.n_heads:

            repeat_factor = (
                self.n_heads
                // self.n_kv_heads
            )

            k = k.repeat_interleave(
                repeat_factor,
                dim=1,
            )

            v = v.repeat_interleave(
                repeat_factor,
                dim=1,
            )

        # ====================================================
        # Flash / SDPA Attention
        # ====================================================

        # During cached decoding, only one new token is
        # queried, so causal masking is unnecessary.

        is_causal = (
            self.causal
            and kv is None
            and past_kv is None
        )

        attention_mask = None
        if self.causal and kv is None and past_kv is not None:
            offset = past_kv[0].size(2)
            attention_mask = (torch.arange(k.size(2), device=x.device)[None, :] <=
                              (offset + torch.arange(T, device=x.device))[:, None])[None, None]
        if key_mask is not None:
            valid_keys = key_mask[:, None, None, :].bool()
            attention_mask = valid_keys if attention_mask is None else attention_mask & valid_keys
            if is_causal:
                causal_mask = torch.ones(T, k.size(2), device=x.device, dtype=torch.bool).tril()
                attention_mask = attention_mask & causal_mask[None, None]
            is_causal = False

        out = F.scaled_dot_product_attention(
            q,
            k,
            v,
            attn_mask=attention_mask,
            dropout_p=0.0,
            is_causal=is_causal,
        )

        # ====================================================
        # Output projection
        # ====================================================

        out = out.transpose(
            1,
            2,
        ).contiguous()

        out = out.view(
            B,
            T,
            C,
        )

        out = self.o_proj(out)

        # Cache only decoder self-attention
        new_kv = None

        if use_cache and kv is None:
            new_kv = cache_kv

        return out, new_kv


# ============================================================
# Encoder Block
# ============================================================

class EncoderBlock(nn.Module):

    def __init__(
        self,
        dim,
        n_heads,
        n_kv_heads,
        ff_dim,
        max_seq_len,
    ):
        super().__init__()

        self.norm1 = RMSNorm(dim)

        self.self_attn = Attention(
            dim=dim,
            n_heads=n_heads,
            n_kv_heads=n_kv_heads,
            max_seq_len=max_seq_len,
            causal=False,
            use_rope=True,
        )

        self.norm2 = RMSNorm(dim)

        self.ffn = SwiGLU(
            dim,
            ff_dim,
        )

    def forward(self, x, key_mask=None):

        # Pre-norm self-attention
        x = x + self.self_attn(
            self.norm1(x), key_mask=key_mask
        )[0]

        # Pre-norm FFN
        x = x + self.ffn(
            self.norm2(x)
        )

        return x


# ============================================================
# Decoder Block
# ============================================================

class DecoderBlock(nn.Module):

    def __init__(
        self,
        dim,
        n_heads,
        n_kv_heads,
        ff_dim,
        max_self_attn_len,
        max_cross_attn_len,
    ):
        super().__init__()

        # ----------------------------------------------------
        # Causal self-attention
        # ----------------------------------------------------

        self.norm1 = RMSNorm(dim)

        self.self_attn = Attention(
            dim=dim,
            n_heads=n_heads,
            n_kv_heads=n_kv_heads,
            max_seq_len=max_self_attn_len,
            causal=True,
            use_rope=True,
        )

        # ----------------------------------------------------
        # Encoder-decoder cross-attention
        # ----------------------------------------------------

        self.norm2 = RMSNorm(dim)

        self.cross_attn = Attention(
            dim=dim,
            n_heads=n_heads,
            n_kv_heads=n_kv_heads,
            max_seq_len=max_cross_attn_len,
            causal=False,
            use_rope=False,
        )

        # ----------------------------------------------------
        # FFN
        # ----------------------------------------------------

        self.norm3 = RMSNorm(dim)

        self.ffn = SwiGLU(
            dim,
            ff_dim,
        )

    def forward(
        self,
        x,
        encoder_output,
        past_kv=None,
        use_cache=False,
        key_mask=None,
        encoder_mask=None,
    ):

        # ====================================================
        # 1. Decoder causal self-attention
        # ====================================================

        residual = x

        x_norm = self.norm1(x)

        self_out, new_kv = self.self_attn(
            x_norm,
            past_kv=past_kv,
            use_cache=use_cache,
            key_mask=key_mask,
        )

        x = residual + self_out

        # ====================================================
        # 2. Cross-attention
        # ====================================================

        residual = x

        x_norm = self.norm2(x)

        cross_out, _ = self.cross_attn(
            x_norm,
            kv=encoder_output,
            key_mask=encoder_mask,
        )

        x = residual + cross_out

        # ====================================================
        # 3. FFN
        # ====================================================

        x = x + self.ffn(
            self.norm3(x)
        )

        return x, new_kv


# ============================================================
# Encoder-Decoder Language Model
# ============================================================

class EncoderDecoderLM(nn.Module):

    def __init__(
        self,
        vocab_size=VOCAB_SIZE,
        dim=DIM,
        encoder_layers=ENCODER_LAYERS,
        decoder_layers=DECODER_LAYERS,
        n_heads=N_HEADS,
        n_kv_heads=N_KV_HEADS,
        ff_dim=FF_DIM,
        max_encoder_len=MAX_ENCODER_LEN,
        max_decoder_len=MAX_DECODER_LEN,
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.dim = dim

        # ====================================================
        # Shared token embedding
        # ====================================================

        self.token_embedding = nn.Embedding(
            vocab_size,
            dim,
        )

        # ====================================================
        # Encoder
        # ====================================================

        self.encoder_blocks = nn.ModuleList([
            EncoderBlock(
                dim=dim,
                n_heads=n_heads,
                n_kv_heads=n_kv_heads,
                ff_dim=ff_dim,
                max_seq_len=max_encoder_len,
            )
            for _ in range(encoder_layers)
        ])

        self.encoder_norm = RMSNorm(dim)

        # ====================================================
        # Decoder
        # ====================================================

        self.decoder_blocks = nn.ModuleList([
            DecoderBlock(
                dim=dim,
                n_heads=n_heads,
                n_kv_heads=n_kv_heads,
                ff_dim=ff_dim,
                max_self_attn_len=max_decoder_len,
                max_cross_attn_len=max_encoder_len,
            )
            for _ in range(decoder_layers)
        ])

        self.decoder_norm = RMSNorm(dim)

        # ====================================================
        # Output head
        # ====================================================

        self.lm_head = nn.Linear(
            dim,
            vocab_size,
            bias=False,
        )

        # Tie output projection to token embedding
        self.lm_head.weight = (
            self.token_embedding.weight
        )

    # ========================================================
    # ENCODER
    # ========================================================

    def encode(self, encoder_ids):

        x = self.token_embedding(
            encoder_ids
        )

        for block in self.encoder_blocks:
            x = block(x)

        x = self.encoder_norm(x)

        return x

    # ========================================================
    # DECODER
    # ========================================================

    def decode(
        self,
        decoder_ids,
        encoder_output,
        past_kv=None,
        use_cache=False,
    ):

        x = self.token_embedding(
            decoder_ids
        )

        new_cache = []

        for i, block in enumerate(
            self.decoder_blocks
        ):

            layer_past = None

            if past_kv is not None:
                layer_past = past_kv[i]

            x, layer_cache = block(
                x=x,
                encoder_output=encoder_output,
                past_kv=layer_past,
                use_cache=use_cache,
            )

            if use_cache:
                new_cache.append(
                    layer_cache
                )

        x = self.decoder_norm(x)

        logits = self.lm_head(x)

        if use_cache:
            return logits, new_cache

        return logits, None

    # ========================================================
    # FORWARD
    # ========================================================

    def forward(
        self,
        encoder_ids,
        decoder_ids,
        labels=None,
    ):

        # Encode 512-token context
        encoder_output = self.encode(
            encoder_ids
        )

        # Decode continuation
        logits, _ = self.decode(
            decoder_ids,
            encoder_output,
        )

        # No labels -> inference
        if labels is None:
            return logits

        # ====================================================
        # Language-model loss
        # ====================================================

        loss = F.cross_entropy(
            logits.reshape(
                -1,
                self.vocab_size,
            ),
            labels.reshape(-1),
            ignore_index=-100,
        )

        return logits, loss

    # ========================================================
    # GENERATION
    # ========================================================

    @torch.no_grad()
    def generate(
        self,
        encoder_ids,
        bos_id,
        eos_id=None,
        max_new_tokens=128,
        temperature=1.0,
        top_p=0.9,
    ):

        self.eval()

        # ----------------------------------------------------
        # Encode context ONCE
        # ----------------------------------------------------

        encoder_output = self.encode(
            encoder_ids
        )

        # ----------------------------------------------------
        # Start decoder
        # ----------------------------------------------------

        B = encoder_ids.size(0)

        decoder_ids = torch.full(
            (B, 1),
            bos_id,
            dtype=torch.long,
            device=encoder_ids.device,
        )

        past_kv = None

        # ----------------------------------------------------
        # Autoregressive generation
        # ----------------------------------------------------

        for step in range(max_new_tokens):

            # Only feed the newest token once cache exists
            if past_kv is None:
                current_ids = decoder_ids
            else:
                current_ids = decoder_ids[:, -1:]

            logits, past_kv = self.decode(
                decoder_ids=current_ids,
                encoder_output=encoder_output,
                past_kv=past_kv,
                use_cache=True,
            )

            next_logits = logits[:, -1, :]

            # Temperature
            if temperature != 1.0:
                next_logits = (
                    next_logits / temperature
                )

            # ------------------------------------------------
            # Top-p sampling
            # ------------------------------------------------

            probs = F.softmax(
                next_logits,
                dim=-1,
            )

            sorted_probs, sorted_indices = torch.sort(
                probs,
                descending=True,
                dim=-1,
            )

            cumulative_probs = torch.cumsum(
                sorted_probs,
                dim=-1,
            )

            remove = (
                cumulative_probs > top_p
            )

            # Keep at least the first token
            remove[:, 0] = False

            sorted_probs[remove] = 0.0

            sorted_probs = (
                sorted_probs
                /
                sorted_probs.sum(
                    dim=-1,
                    keepdim=True,
                )
            )

            next_token = torch.multinomial(
                sorted_probs,
                num_samples=1,
            )

            next_token = torch.gather(
                sorted_indices,
                -1,
                next_token,
            )

            decoder_ids = torch.cat(
                [
                    decoder_ids,
                    next_token,
                ],
                dim=1,
            )

            if (
                eos_id is not None
                and torch.all(
                    next_token == eos_id
                )
            ):
                break

        # Remove BOS
        return decoder_ids[:, 1:]


# ============================================================
# Parameter count
# ============================================================


```

```python
def memory_for_active_precision(memory, model):
    """Match the active execution context, not the global AMP preference.

    Startup checks run outside autocast even on an AMP-enabled GPU. In that
    case projections require the model weight dtype. Tensor.to preserves the
    gradient connection needed by CPU representation replay.
    """
    if memory is None:
        return None
    weight = model.token_embedding.weight
    dtype = (torch.get_autocast_dtype(weight.device.type)
             if torch.is_autocast_enabled(weight.device.type) else weight.dtype)
    return memory.to(dtype=dtype)

class LoRMALinear(nn.Module):
    def __init__(self, original_linear, r=16, alpha=32, dropout=0.05, do_pi=True):
        super().__init__()
        self.original_linear = original_linear
        self.in_features = original_linear.in_features
        self.out_features = original_linear.out_features
        self.scaling, self.do_pi = alpha / r, do_pi
        self.dropout = nn.Dropout(dropout)
        for p in original_linear.parameters(): p.requires_grad = False
        W = original_linear.weight
        self.A = nn.Parameter(W.new_empty(r, self.out_features))
        self.B = nn.Parameter(W.new_zeros(self.out_features, r))
        nn.init.kaiming_uniform_(self.A, a=math.sqrt(5))
    def effective_weight(self):
        W = self.original_linear.weight
        delta = self.B @ self.A
        if self.do_pi:
            rows = torch.arange(self.out_features, device=W.device)[:, None]
            cols = torch.arange(self.out_features, device=W.device)[None, :]
            delta = delta.gather(1, (cols - rows) % self.out_features)
        return W + self.scaling * (delta @ W)
    def forward(self, x):
        W = self.original_linear.weight
        W_eff = self.effective_weight()
        return F.linear(x, W, self.original_linear.bias) + F.linear(self.dropout(x), W_eff - W)


from torch.utils.checkpoint import checkpoint as activation_checkpoint
from contextlib import contextmanager

class SummaryModel(EncoderDecoderLM):
    def initialize_context_modules(self):
        # Copy values into separate modules; never alias original parameter storage.
        self.context_encoder = EncoderBlock(self.dim, MODEL_CONFIG['n_heads'], MODEL_CONFIG['n_kv_heads'],
                                            MODEL_CONFIG['ff_dim'], MAX_SEQ_LEN)
        self.context_encoder.load_state_dict(self.encoder_blocks[-1].state_dict(), strict=True)
        self.context_encoder_norm = RMSNorm(self.dim)
        self.context_encoder_norm.load_state_dict(self.encoder_norm.state_dict(), strict=True)
        self.context_attention = Attention(self.dim, MODEL_CONFIG['n_heads'], MODEL_CONFIG['n_kv_heads'],
                                           MAX_SEQ_LEN, causal=False, use_rope=False)
        self.context_attention.load_state_dict(self.decoder_blocks[-1].cross_attn.state_dict(), strict=True)
        self.context_query_norm = RMSNorm(self.dim)
        self.context_query_norm.load_state_dict(self.decoder_norm.state_dict(), strict=True)
        self.context_gate = nn.Parameter(torch.tensor(CONTEXT_GATE_INIT))

    def encode_chunk(self, ids, mask, context_only=False):
        x = self.token_embedding(ids)
        blocks = [self.context_encoder] if context_only else self.encoder_blocks
        for block in blocks:
            if self.training and USE_GRADIENT_CHECKPOINTING:
                x = activation_checkpoint(block,x,key_mask=mask,use_reentrant=False,preserve_rng_state=True)
            else: x = block(x,key_mask=mask)
        return self.context_encoder_norm(x) if context_only else self.encoder_norm(x)

    def chunk_position(self, index, reference):
        # Non-trainable source-order signal; no new external pretrained weights.
        frequencies = torch.exp(torch.arange(0,self.dim,2,device=reference.device).float()*(-math.log(10000.0)/self.dim))
        angle = index * frequencies
        signal = torch.stack((angle.sin(),angle.cos()),dim=-1).flatten()[:self.dim]
        return signal.to(reference.dtype)[None,:] * CHUNK_POSITION_SCALE

    def pool_tokens(self,h,limit):
        # Compress contiguous positions; every token contributes to a pooling bin.
        if h.size(0)<=limit: return h
        return F.adaptive_avg_pool1d(h.transpose(0,1)[None],limit)[0].transpose(0,1)

    def encode_and_pool(self,ids,mask,context_only=False):
        h = self.encode_chunk(ids,mask,context_only=context_only)
        limit = CONTEXT_TOKENS_PER_CHUNK if context_only else MAIN_TOKENS_PER_CHUNK
        # Microbatch is one unpadded source region. Pool inside the checkpointed
        # region, so full-resolution encoder outputs are not retained per chunk.
        return self.pool_tokens(h[0],limit)

    def encode_document(self,chunks):
        device = self.token_embedding.weight.device
        main_parts,context_parts = [],[]
        def encode_region(tokens,context_only):
            ids = torch.tensor([tokens],device=device,dtype=torch.long)
            mask = torch.ones_like(ids,dtype=torch.bool)
            if self.training and USE_GRADIENT_CHECKPOINTING:
                h = activation_checkpoint(self.encode_and_pool,ids,mask,context_only=context_only,
                    use_reentrant=False,preserve_rng_state=True)
            else: h = self.encode_and_pool(ids,mask,context_only=context_only)
            return h
        for i,chunk in enumerate(chunks):
            if not chunk or len(chunk)>MAX_SEQ_LEN: raise ValueError('Invalid current chunk size')
            h = encode_region(chunk,False)
            main_parts.append(h+self.chunk_position(i,h))
            context_tokens = neighbor_context_tokens(chunks,i)
            if context_tokens:
                h_context = encode_region(context_tokens,True)
                context_parts.append(h_context+self.chunk_position(i,h_context))
        main = self.pool_tokens(torch.cat(main_parts,dim=0),MAX_MAIN_MEMORY_TOKENS)[None]
        context = (self.pool_tokens(torch.cat(context_parts,dim=0),MAX_CONTEXT_MEMORY_TOKENS)[None]
                   if context_parts else None)
        main = memory_for_active_precision(main,self)
        context = memory_for_active_precision(context,self)
        return main,context

    def decode(self, decoder_ids, encoder_output, past_kv=None, use_cache=False,
               decoder_mask=None, encoder_mask=None, context_memory=None):
        x = self.token_embedding(decoder_ids)
        new_cache = []
        for i,block in enumerate(self.decoder_blocks):
            layer_past = None if past_kv is None else past_kv[i]
            if self.training and USE_GRADIENT_CHECKPOINTING:
                if use_cache: raise ValueError('Training does not use the inference cache')
                x,layer_cache = activation_checkpoint(block,x,encoder_output,key_mask=decoder_mask,
                    encoder_mask=encoder_mask,use_reentrant=False,preserve_rng_state=True)
            else:
                x,layer_cache = block(x,encoder_output,past_kv=layer_past,use_cache=use_cache,
                                      key_mask=decoder_mask,encoder_mask=encoder_mask)
            if use_cache: new_cache.append(layer_cache)
        h = self.decoder_norm(x)
        if context_memory is not None:
            if self.training and USE_GRADIENT_CHECKPOINTING:
                delta = activation_checkpoint(self.context_residual,h,context_memory,
                                              use_reentrant=False,preserve_rng_state=True)
            else: delta = self.context_residual(h,context_memory)
            h = h + self.context_gate.sigmoid()*delta
        return self.lm_head(h),new_cache if use_cache else None

    def context_residual(self,h,memory):
        return self.context_attention(self.context_query_norm(h),kv=memory)[0]

    def forward(self,chunks,decoder_ids,decoder_mask=None):
        main,context = self.encode_document(chunks)
        logits,_ = self.decode(decoder_ids,main,decoder_mask=decoder_mask,context_memory=context)
        return logits

phase1_path = download_bucket_file(MODEL_PATH)
TEACHER = None
TEACHER_INFO = None

def evenly_spaced_layers(original_count, student_count):
    if student_count == 1: return [original_count-1]
    return [round(i*(original_count-1)/(student_count-1)) for i in range(student_count)]

ENCODER_LAYER_MAP = evenly_spaced_layers(PHASE1_MODEL_CONFIG['encoder_layers'],STUDENT_ENCODER_LAYERS)
DECODER_LAYER_MAP = evenly_spaced_layers(PHASE1_MODEL_CONFIG['decoder_layers'],STUDENT_DECODER_LAYERS)

def extract_model_state(checkpoint):
    if 'model_state_dict' in checkpoint: return dict(checkpoint['model_state_dict'])
    if 'model' in checkpoint: return dict(checkpoint['model'])
    if 'state_dict' in checkpoint: return dict(checkpoint['state_dict'])
    if all(isinstance(v,torch.Tensor) for v in checkpoint.values()): return dict(checkpoint)
    raise ValueError(f'Cannot identify model weights: {list(checkpoint)}')

def build_summary_model():
    # Remap retained Phase 1 layers into the smaller, dense student architecture.
    checkpoint = torch.load(phase1_path,map_location='cpu',weights_only=False)
    foundation = extract_model_state(checkpoint)
    model = SummaryModel(vocab_size=VOCAB_SIZE,**MODEL_CONFIG)
    mapped = {}
    for key in model.state_dict():
        source_key = key
        for stack, mapping in [('encoder_blocks',ENCODER_LAYER_MAP),('decoder_blocks',DECODER_LAYER_MAP)]:
            if key.startswith(stack+'.'):
                parts = key.split('.')
                parts[1] = str(mapping[int(parts[1])])
                source_key = '.'.join(parts)
        if source_key == 'lm_head.weight' and source_key not in foundation: source_key = 'token_embedding.weight'
        if source_key not in foundation: raise ValueError(f'Missing Phase 1 tensor: {source_key}')
        mapped[key] = foundation[source_key]
    if not torch.equal(mapped['lm_head.weight'],mapped['token_embedding.weight']):
        raise ValueError('Phase 1 checkpoint does not match the tied-head architecture')
    model.load_state_dict(mapped,strict=True)
    del foundation,mapped,checkpoint
    model.initialize_context_modules()
    for parameter in model.parameters(): parameter.requires_grad = True
    model = model.to(DEVICE)
    assert model.lm_head.weight is model.token_embedding.weight
    print('Retained Phase 1 encoder/decoder indices:',ENCODER_LAYER_MAP,DECODER_LAYER_MAP)
    print('Student total/trainable parameters:',sum(p.numel() for p in model.parameters()),
          sum(p.numel() for p in model.parameters() if p.requires_grad))
    return model

def teacher_checkpoint_remote():
    if TEACHER_CHECKPOINT_PATH: return bucket_path(TEACHER_CHECKPOINT_PATH)
    if TEACHER_TAG: return checkpoint_remote_path(TEACHER_TAG)
    candidates = sorted(fs.glob(bucket_path(f"{CHECKPOINT_PREFIX.strip('/')}/*/checkpoint.pt")))
    candidates = [p for p in candidates if 'encoder_decoder' in p and 'lorma' in p.lower()
                  and p.rstrip('/').split('/')[-2].endswith('_best')]
    if len(candidates) != 1:
        print('Available encoder-decoder LORMA best checkpoint paths:',candidates)
        raise ValueError('Set TEACHER_TAG or TEACHER_CHECKPOINT_PATH to the exact completed teacher checkpoint. Expected exactly one candidate.')
    return candidates[0] if candidates[0].startswith('hf://') else 'hf://'+candidates[0]

def file_sha256(path):
    h = hashlib.sha256()
    with open(path,'rb') as f:
        for block in iter(lambda:f.read(8*1024*1024),b''): h.update(block)
    return h.hexdigest()

@torch.no_grad()
def merge_teacher_lorma(model):
    # Traverse registered modules once; replace only adapters, preserving all other tensors.
    for parent in list(model.modules()):
        for name,layer in list(parent.named_children()):
            if isinstance(layer,LoRMALinear):
                w = layer.effective_weight().detach()
                merged = nn.Linear(layer.in_features,layer.out_features,
                    bias=layer.original_linear.bias is not None,device=w.device,dtype=w.dtype)
                merged.weight.copy_(w)
                if merged.bias is not None: merged.bias.copy_(layer.original_linear.bias)
                merged.requires_grad_(False)
                setattr(parent,name,merged)
    return model

def build_teacher():
    global TEACHER_INFO
    remote = teacher_checkpoint_remote()
    local = download_bucket_file(remote)
    digest = file_sha256(local)
    payload = torch.load(local,map_location='cpu',weights_only=False)
    metadata = payload.get('metadata',{})
    if metadata.get('domain') != 'SUMM' or metadata.get('phase1_tag') != PHASE1_TAG:
        raise ValueError('Teacher must be a Phase 6 encoder-decoder checkpoint from this Phase 1 architecture')
    if metadata.get('model_config') != PHASE1_MODEL_CONFIG or metadata.get('tokenizer_path') != TOKENIZER_PATH:
        raise ValueError('Teacher architecture/tokenizer differ from this notebook')
    if 'lorma' not in str(metadata.get('technique','')).lower(): raise ValueError('Expected a trained LORMA teacher')
    expected = dict(data_fingerprint=DATA_FINGERPRINT,seed=SEED,val_fraction=VAL_FRACTION,
                    val_max_docs=VAL_MAX_DOCS,train_fraction=TRAIN_FRACTION)
    mismatches = {k:(metadata.get(k),v) for k,v in expected.items() if metadata.get(k)!=v}
    if mismatches: raise ValueError(f'Teacher dataset/split settings mismatch (potential validation leakage): {mismatches}')
    if metadata.get('max_new_tokens') != MAX_NEW_TOKENS:
        raise ValueError('Teacher and student target truncation budgets must match')
    if metadata.get('source_selection')=='all_consecutive_chunks':
        if metadata.get('chunk_position_scale')!=CHUNK_POSITION_SCALE:
            raise ValueError('Context teacher chunk-position scale differs; restore the original configuration')
    selection = metadata.get('source_selection')
    if selection != 'uniform_spans_with_gaps_v1':
        raise ValueError(f'Use the original trained 512-token LORMA teacher for this Kaggle run, not the all-chunks teacher. Got: {selection}')
    context_teacher = selection=='all_consecutive_chunks'
    teacher = SummaryModel(vocab_size=VOCAB_SIZE,**PHASE1_MODEL_CONFIG)
    if context_teacher: teacher.initialize_context_modules()
    count = metadata.get('lorma_last_n_blocks',TEACHER_LORMA_LAST_N_BLOCKS)
    attention_modules = [b.self_attn for b in teacher.encoder_blocks[-count:]]
    attention_modules += [a for b in teacher.decoder_blocks[-count:] for a in (b.self_attn,b.cross_attn)]
    if context_teacher: attention_modules += [teacher.context_encoder.self_attn,teacher.context_attention]
    for attention in attention_modules:
        for name in ['q_proj','k_proj','v_proj','o_proj']:
            setattr(attention,name,LoRMALinear(getattr(attention,name),**metadata['adapter_config']))
    teacher.load_state_dict(payload['model'],strict=True)
    teacher_epoch = payload.get('epoch')
    del payload
    teacher = merge_teacher_lorma(teacher)
    teacher.requires_grad_(False)
    teacher.eval()
    teacher.teacher_metadata = metadata
    TEACHER_INFO = dict(remote_path=remote,sha256=digest,epoch=teacher_epoch,
                        source_selection=selection,metadata=metadata,
                        parameters=sum(p.numel() for p in teacher.parameters()))
    print('Teacher checkpoint:',remote,'| source strategy:',selection,'| parameters:',TEACHER_INFO['parameters'])
    print('Frozen teacher device:',TEACHER_DEVICE,'| student device:',DEVICE)
    return teacher.to(TEACHER_DEVICE)

def select_teacher_source(tokens,metadata):
    budget = metadata['source_budget']
    spans = metadata['source_spans']
    if len(tokens)<=budget: return tokens or [EOS_ID]
    gap = tokenizer.encode('\n[...]\n',add_special_tokens=False).ids
    spans = min(spans,budget)
    content_budget = budget-(spans-1)*len(gap)
    if content_budget<spans: raise ValueError('Invalid teacher selection budget')
    if spans==1: return tokens[:budget]
    widths = [content_budget//spans+int(i<content_budget%spans) for i in range(spans)]
    omitted,consumed,selected = len(tokens)-content_budget,0,[]
    for i,width in enumerate(widths):
        start = consumed+i*omitted//(spans-1)
        if i: selected.extend(gap)
        selected.extend(tokens[start:start+width]);consumed += width
    assert len(selected)==budget
    return selected

@torch.no_grad()
def teacher_memories(teacher,chunks):
    metadata = teacher.teacher_metadata
    if metadata['source_selection']=='all_consecutive_chunks':
        return teacher.encode_document(chunks)
    tokens = [token for chunk in chunks for token in chunk]
    selected = select_teacher_source(tokens,metadata)
    ids = torch.tensor([selected],device=next(teacher.parameters()).device,dtype=torch.long)
    mask = torch.ones_like(ids,dtype=torch.bool)
    return teacher.encode_chunk(ids,mask),None

@torch.no_grad()
def teacher_logits(teacher,batch):
    teacher.eval()
    with autocast_context():
        main,context = teacher_memories(teacher,batch['chunks'])
        logits,_ = teacher.decode(batch['decoder_ids'].to(next(teacher.parameters()).device),main,
            decoder_mask=batch['decoder_mask'].to(next(teacher.parameters()).device),context_memory=context)
    return logits.detach().to(DEVICE)

def masked_distillation_terms(student_logits,teacher_logits_tensor,labels):
    valid = labels.reshape(-1)!=-100
    student = student_logits.reshape(-1,VOCAB_SIZE)[valid]
    target = labels.reshape(-1)[valid]
    count = int(valid.sum())
    ce_sum = student.new_zeros((),dtype=torch.float32)
    kl_sum = student.new_zeros((),dtype=torch.float32)
    teacher_flat = None if teacher_logits_tensor is None else teacher_logits_tensor.reshape(-1,VOCAB_SIZE)[valid]
    for start in range(0,count,KD_TOKEN_MICROBATCH):
        end = start+KD_TOKEN_MICROBATCH
        s = student[start:end].float()
        ce_sum = ce_sum + F.cross_entropy(s,target[start:end],reduction='sum')
        if teacher_flat is not None:
            teacher_log = F.log_softmax(teacher_flat[start:end].float()/KD_TEMPERATURE,dim=-1)
            student_log = F.log_softmax(s/KD_TEMPERATURE,dim=-1)
            kl_sum = kl_sum + F.kl_div(student_log,teacher_log,log_target=True,reduction='sum') * KD_TEMPERATURE**2
    return ce_sum,kl_sum,count

def summary_loss_sum(model,batch):
    decoder_ids = batch['decoder_ids'].to(DEVICE)
    decoder_mask = batch['decoder_mask'].to(DEVICE)
    labels = batch['labels'].to(DEVICE)
    # Teacher is used only during student training; validation monitors reference CE.
    target_logits = teacher_logits(TEACHER,batch) if model.training else None
    with autocast_context(): student_logits = model(batch['chunks'],decoder_ids,decoder_mask)
    ce,kl,count = masked_distillation_terms(student_logits,target_logits,labels)
    objective = (1-KD_WEIGHT)*ce + KD_WEIGHT*kl if model.training else ce
    if model.training: model.last_kd_terms = dict(reference_ce=ce.detach().item()/count,
                                                  teacher_kl=kl.detach().item()/count)
    return objective,count

@torch.no_grad()
def generate_summary(model,chunks):
    model.eval()
    with autocast_context(): main,context = model.encode_document(chunks)
    ids = torch.tensor([[DECODER_START_ID]],device=DEVICE,dtype=torch.long)
    past,generated = None,[]
    for _ in range(MAX_NEW_TOKENS):
        with autocast_context():
            logits,past = model.decode(ids,main,past_kv=past,use_cache=True,context_memory=context)
        next_id = int(logits[0,-1].argmax())
        if next_id==EOS_ID: break
        generated.append(next_id)
        ids = torch.tensor([[next_id]],device=DEVICE,dtype=torch.long)
    return tokenizer.decode(generated,skip_special_tokens=True).strip(),len(generated)

@contextmanager
def merged_lorma_for_generation(model):
    # Materialize effective weights once, rather than on every generated token.
    # Restore the original modules (and optimizer parameter identities) afterwards.
    if model.training: raise RuntimeError('Merge is only for evaluation')
    originals = []
    try:
        with torch.no_grad():
            attention_modules = [b.self_attn for b in model.encoder_blocks]
            attention_modules += [a for b in model.decoder_blocks for a in (b.self_attn,b.cross_attn)]
            attention_modules += [model.context_encoder.self_attn, model.context_attention]
            for attention in attention_modules:
                for name in ['q_proj','k_proj','v_proj','o_proj']:
                    layer = getattr(attention,name)
                    if isinstance(layer,LoRMALinear):
                        weight = layer.effective_weight().detach()
                        merged = nn.Linear(layer.in_features,layer.out_features,bias=layer.original_linear.bias is not None,
                                           device=weight.device,dtype=weight.dtype)
                        merged.weight.copy_(weight)
                        if merged.bias is not None: merged.bias.copy_(layer.original_linear.bias)
                        merged.requires_grad_(False)
                        merged.eval()
                        originals.append((attention,name,layer))
                        setattr(attention,name,merged)
        yield
    finally:
        for attention,name,layer in originals: setattr(attention,name,layer)

@torch.no_grad()
def check_generation_cache(model):
    model.eval()
    # Two synthetic short chunks exercise context and a memory longer than one chunk.
    row = datasets['train'].rows[0]
    first = row['chunks'][0][:12]
    chunks = [first,first]
    decoder = torch.tensor([[DECODER_START_ID]+row['target'][:2]],device=DEVICE)
    with merged_lorma_for_generation(model):
        main,context = model.encode_document(chunks)
        expected,_ = model.decode(decoder,main,context_memory=context)
        past = None
        for i in range(decoder.size(1)):
            actual,past = model.decode(decoder[:,i:i+1],main,past_kv=past,use_cache=True,context_memory=context)
            torch.testing.assert_close(actual,expected[:,i:i+1],rtol=2e-4,atol=2e-4)
    print('Document/context cached and full decoder logits agree on startup check')

def check_document_gradients(model):
    # Short synthetic multi-chunk input verifies the complete training graph
    # before starting the long run; does not perform an optimizer update.
    was_training = model.training
    model.train()
    model.zero_grad(set_to_none=True)
    try:
        row = datasets['train'].rows[0]
        source = row['chunks'][0][:12]
        target = row['target'][:3]
        ids = torch.tensor([[DECODER_START_ID]+target[:-1]],device=DEVICE)
        labels = torch.tensor([target],device=DEVICE)
        with autocast_context(): logits = model([source,source[::-1]],ids)
        loss = F.cross_entropy(logits.float().reshape(-1,VOCAB_SIZE),labels.reshape(-1))
        loss.backward()
        for name, parameter in [('main encoder',model.encoder_blocks[-1].self_attn.v_proj.weight),
                                ('context encoder',model.context_encoder.self_attn.v_proj.weight),
                                ('late context attention',model.context_attention.v_proj.weight),
                                ('context gate',model.context_gate)]:
            grad = parameter.grad
            if grad is None or not torch.isfinite(grad).all() or not (grad.abs().sum()>0):
                raise RuntimeError(f'Missing/nonfinite/zero startup gradient in {name}')
        print('Whole-document loss reaches main encoder, context encoder, late attention and gate')
    finally:
        model.zero_grad(set_to_none=True)
        model.train(was_training)

@torch.no_grad()
def check_distillation_terms(model):
    batch = next(iter(make_loader('dev')))
    labels = batch['labels'].to(DEVICE)
    t = teacher_logits(TEACHER,batch)
    ce,kl,count = masked_distillation_terms(t,t,labels)
    if not torch.isfinite(ce) or not torch.isfinite(kl) or abs(kl.item()/count)>1e-5:
        raise RuntimeError('Distillation self-KL check failed')
    assert not any(p.requires_grad for p in TEACHER.parameters())
    assert not t.requires_grad and TEACHER.training is False
    print('Teacher frozen; masked self-KL and finite reference loss checks passed')

@torch.no_grad()
def generate_teacher_summary(teacher,chunks):
    with autocast_context(): main,context = teacher_memories(teacher,chunks)
    ids = torch.tensor([[DECODER_START_ID]],device=next(teacher.parameters()).device,dtype=torch.long)
    past,generated = None,[]
    for _ in range(MAX_NEW_TOKENS):
        with autocast_context():
            logits,past = teacher.decode(ids,main,past_kv=past,use_cache=True,context_memory=context)
        token = int(logits[0,-1].argmax())
        if token==EOS_ID: break
        generated.append(token)
        ids = torch.tensor([[token]],device=next(teacher.parameters()).device,dtype=torch.long)
    return tokenizer.decode(generated,skip_special_tokens=True).strip(),len(generated)

@torch.no_grad()
def compare_teacher_student(student,student_metrics):
    scorer = rouge_scorer.RougeScorer(['rouge1','rougeL'],use_stemmer=True)
    totals = dict(rouge1_f1=0.0,rougeL_f1=0.0,generated_tokens=0.0)
    started = time.perf_counter()
    for row in tqdm(datasets['dev'].rows,desc='Teacher validation comparison'):
        summary,count = generate_teacher_summary(TEACHER,row['chunks'])
        scores = scorer.score(row['reference'],summary)
        totals['rouge1_f1'] += scores['rouge1'].fmeasure
        totals['rougeL_f1'] += scores['rougeL'].fmeasure
        totals['generated_tokens'] += count
    teacher_metrics = {k:v/len(datasets['dev']) for k,v in totals.items()}
    teacher_metrics['generation_seconds'] = time.perf_counter()-started
    # End-to-end generation timing: same documents, includes source encoding and decoding.
    timings = {}
    rows = datasets['dev'].rows[:min(5,len(datasets['dev']))]
    for name,model,function in [('teacher',TEACHER,generate_teacher_summary),('student',student,generate_summary)]:
        model.eval()
        function(model,rows[0]['chunks'])  # warm up
        if next(model.parameters()).device.type=='cuda': torch.cuda.synchronize(next(model.parameters()).device)
        started = time.perf_counter()
        for row in rows: function(model,row['chunks'])
        if next(model.parameters()).device.type=='cuda': torch.cuda.synchronize(next(model.parameters()).device)
        timings[name] = dict(documents=len(rows),seconds_per_document=(time.perf_counter()-started)/len(rows),
                             parameters=sum(p.numel() for p in model.parameters()),
                             parameter_bytes=sum(p.numel()*p.element_size() for p in model.parameters()))
    comparison = dict(teacher=teacher_metrics,student=student_metrics,inference=timings,
                      teacher_source_selection=TEACHER_INFO['source_selection'],student_source_selection='all_chunks_bounded_pooled_memory')
    with open(WORK_DIR/'distillation_comparison.json','w') as f: json.dump(comparison,f,indent=2)
    print('Teacher/student validation and student-only inference comparison:',comparison)
    return comparison

def training_objective_from_logits(model,batch,student_logits):
    labels = batch['labels'].to(DEVICE)
    target_logits = teacher_logits(TEACHER,batch)
    ce,kl,count = masked_distillation_terms(student_logits,target_logits,labels)
    model.last_kd_terms = dict(reference_ce=ce.detach().item()/count,teacher_kl=kl.detach().item()/count)
    return (1-KD_WEIGHT)*ce+KD_WEIGHT*kl,count

def capture_region_rng(device):
    state = dict(cpu=torch.get_rng_state())
    if device.type=='cuda': state['cuda'] = torch.cuda.get_rng_state(device)
    return state

@torch.no_grad()
def cache_document_regions(model,chunks):
    # Keep training mode: cache the same dropout samples used during replay.
    # Only CPU representations, token IDs and RNG states survive each region.
    device = model.token_embedding.weight.device
    records = []
    for index,chunk in enumerate(chunks):
        regions = [(False,chunk)]
        context = neighbor_context_tokens(chunks,index)
        if context: regions.append((True,context))
        for context_only,tokens in regions:
            state = capture_region_rng(device)
            ids = torch.tensor([tokens],device=device,dtype=torch.long)
            mask = torch.ones_like(ids,dtype=torch.bool)
            with autocast_context():
                h = model.encode_and_pool(ids,mask,context_only=context_only)
                h = h+model.chunk_position(index,h)
            records.append(dict(context_only=context_only,tokens=tokens,index=index,
                                hidden=h.detach().cpu(),rng=state))
            del h,ids,mask
    return records

def document_memories_from_cpu_cache(model,records):
    device = model.token_embedding.weight.device
    main_cpu = torch.cat([r['hidden'] for r in records if not r['context_only']],dim=0)
    context_parts = [r['hidden'] for r in records if r['context_only']]
    # These leaf tensors form the boundary where document-level gradients are cached.
    main_leaf = main_cpu.to(device).detach().requires_grad_(True)
    context_leaf = (torch.cat(context_parts,dim=0).to(device).detach().requires_grad_(True)
                    if context_parts else None)
    main = model.pool_tokens(main_leaf,MAX_MAIN_MEMORY_TOKENS)[None]
    context = model.pool_tokens(context_leaf,MAX_CONTEXT_MEMORY_TOKENS)[None] if context_leaf is not None else None
    main = memory_for_active_precision(main,model)
    context = memory_for_active_precision(context,model)
    return main,context,main_leaf,context_leaf

def replay_region_encoder_gradients(model,records,main_gradient,context_gradient):
    device = model.token_embedding.weight.device
    cuda_devices = [device.index if device.index is not None else torch.cuda.current_device()] if device.type=='cuda' else []
    main_offset = context_offset = 0
    for record in records:
        length = record['hidden'].size(0)
        if record['context_only']:
            gradient = None if context_gradient is None else context_gradient[context_offset:context_offset+length]
            context_offset += length
        else:
            gradient = None if main_gradient is None else main_gradient[main_offset:main_offset+length]
            main_offset += length
        if gradient is None: continue
        # Replay must use the original region's dropout state. fork_rng restores
        # the enclosing training RNG afterwards, so replay doesn't advance it twice.
        with torch.random.fork_rng(devices=cuda_devices):
            torch.set_rng_state(record['rng']['cpu'])
            if device.type=='cuda': torch.cuda.set_rng_state(record['rng']['cuda'],device)
            ids = torch.tensor([record['tokens']],device=device,dtype=torch.long)
            mask = torch.ones_like(ids,dtype=torch.bool)
            with autocast_context():
                h = model.encode_and_pool(ids,mask,context_only=record['context_only'])
                h = h+model.chunk_position(record['index'],h)
            # Gradient already contains the GradScaler multiplier; do not scale twice.
            torch.autograd.backward(h,gradient.to(device=device,dtype=h.dtype))
            del h,ids,mask

def backward_document_replay(model,batch,scaler,objective=None):
    if not model.training: raise ValueError('Gradient replay is a training operation')
    records = cache_document_regions(model,batch['chunks'])
    main,context,main_leaf,context_leaf = document_memories_from_cpu_cache(model,records)
    decoder_ids = batch['decoder_ids'].to(DEVICE)
    decoder_mask = batch['decoder_mask'].to(DEVICE)
    with autocast_context():
        logits,_ = model.decode(decoder_ids,main,decoder_mask=decoder_mask,context_memory=context)
    if objective is None: loss_sum,count = training_objective_from_logits(model,batch,logits)
    else: loss_sum,count = objective(logits)
    detached_loss = loss_sum.detach()
    # Backward through decoder/head/document pooling, with NO live encoder graphs.
    scaler.scale(loss_sum).backward()
    main_gradient = None if main_leaf.grad is None else main_leaf.grad.detach().cpu()
    context_gradient = None if context_leaf is None or context_leaf.grad is None else context_leaf.grad.detach().cpu()
    # Release decoder activations and GPU representations before encoder replay.
    del loss_sum,logits,main,context,main_leaf,context_leaf,decoder_ids,decoder_mask
    replay_region_encoder_gradients(model,records,main_gradient,context_gradient)
    return detached_loss,count


def check_gradient_replay(model):
    # Compare every trainable parameter gradient with ordinary full autograd on a
    # small two-chunk document. FP32 isolates replay correctness from AMP rounding.
    global USE_AMP
    old_amp,was_training = USE_AMP,model.training
    device = model.token_embedding.weight.device
    cuda_devices = [device.index if device.index is not None else torch.cuda.current_device()] if device.type=='cuda' else []
    try:
        USE_AMP = False
        model.train()
        row = datasets['train'].rows[0]
        source = row['chunks'][0][:12]
        target = row['target'][:3]
        ids = torch.tensor([[DECODER_START_ID]+target[:-1]],device=DEVICE)
        labels = torch.tensor([target],device=DEVICE)
        batch = dict(chunks=[source,source[::-1]],decoder_ids=ids,
                     decoder_mask=torch.ones_like(ids,dtype=torch.bool),labels=labels)
        def objective(logits):
            return F.cross_entropy(logits.float().reshape(-1,VOCAB_SIZE),labels.reshape(-1),reduction='sum'),labels.numel()
        class IdentityScaler:
            def scale(self,loss): return loss
        with torch.random.fork_rng(devices=cuda_devices):
            start_state = capture_region_rng(device)
            model.zero_grad(set_to_none=True)
            reference_loss,_ = objective(model(batch['chunks'],ids,batch['decoder_mask']))
            reference_loss.backward()
            reference = {name:p.grad.detach().cpu().clone() for name,p in model.named_parameters()
                         if p.requires_grad and p.grad is not None}
            reference_value = reference_loss.detach().cpu()
            del reference_loss
            model.zero_grad(set_to_none=True)
            torch.set_rng_state(start_state['cpu'])
            if device.type=='cuda': torch.cuda.set_rng_state(start_state['cuda'],device)
            replay_loss,_ = backward_document_replay(model,batch,IdentityScaler(),objective=objective)
            torch.testing.assert_close(replay_loss.cpu(),reference_value,rtol=5e-4,atol=5e-4)
            for name,p in model.named_parameters():
                if not p.requires_grad: continue
                if name in reference:
                    if p.grad is None: raise RuntimeError(f'Missing replay gradient: {name}')
                    torch.testing.assert_close(p.grad.cpu(),reference[name],rtol=3e-3,atol=3e-4)
                elif p.grad is not None and bool(p.grad.abs().sum()>0):
                    raise RuntimeError(f'Unexpected replay gradient: {name}')
        print('CPU gradient-cache replay agrees with ordinary full-autograd loss and parameter gradients')
    finally:
        model.zero_grad(set_to_none=True)
        USE_AMP = old_amp
        model.train(was_training)
```

```python
def experiment_metadata(domain='SUMM'):
    return dict(domain=domain, phase1_tag=PHASE1_TAG, model_path=MODEL_PATH, tokenizer_path=TOKENIZER_PATH,
                model_config=MODEL_CONFIG, adapter_config=ADAPTER_CONFIG,
                max_seq_len=MAX_SEQ_LEN, max_new_tokens=MAX_NEW_TOKENS, source_chunk_size=MAX_SEQ_LEN,
                source_selection='all_chunks_bounded_pooled_memory', context='previous_tail_next_head_late_attention', context_tokens_per_side=CONTEXT_TOKENS_PER_SIDE,
                main_tokens_per_chunk=MAIN_TOKENS_PER_CHUNK,context_tokens_per_chunk=CONTEXT_TOKENS_PER_CHUNK,
                max_main_memory_tokens=MAX_MAIN_MEMORY_TOKENS,max_context_memory_tokens=MAX_CONTEXT_MEMORY_TOKENS,
                checkpoint_whole_encoder_chunk=True,encoder_backward='cpu_representation_gradient_cache_replay_v1',
                chunk_encode_batch_size=CHUNK_ENCODE_BATCH_SIZE, context_gate_init=CONTEXT_GATE_INIT, chunk_position_scale=CHUNK_POSITION_SCALE, decoder_start_id=DECODER_START_ID, attention='bidirectional_encoder_causal_decoder',
                lorma_last_n_blocks=LORMA_LAST_N_BLOCKS, gradient_checkpointing=USE_GRADIENT_CHECKPOINTING,
                use_amp=USE_AMP, amp_dtype=str(AMP_DTYPE), data_repo=DATA_REPO, data_prefix=DATA_PREFIX,
                data_fingerprint=DATA_FINGERPRINT, val_fraction=VAL_FRACTION, val_max_docs=VAL_MAX_DOCS,
                technique='knowledge_distillation_dense_student', phase1_model_config=PHASE1_MODEL_CONFIG,
                encoder_layer_map=ENCODER_LAYER_MAP, decoder_layer_map=DECODER_LAYER_MAP,
                kd_weight=KD_WEIGHT, kd_temperature=KD_TEMPERATURE, kd_token_microbatch=KD_TOKEN_MICROBATCH, teacher=TEACHER_INFO, train_fraction=TRAIN_FRACTION,
                seed=SEED, batch_size=BATCH_SIZE, grad_accum_steps=GRAD_ACCUM_STEPS,
                lr=LR, weight_decay=WEIGHT_DECAY, grad_clip=GRAD_CLIP, patience=PATIENCE, min_delta=MIN_DELTA,
                scheduler=dict(name='warmup_then_validation_plateau', warmup_ratio=WARMUP_RATIO,
                               start_factor=WARMUP_START_FACTOR, factor=LR_FACTOR, bad_checks=LR_BAD_CHECKS,
                               cooldown_checks=LR_COOLDOWN_CHECKS, min_lr=LR_MIN, loss_threshold=LR_LOSS_THRESHOLD,
                               checks_per_epoch=VAL_LOSS_CHECKS_PER_EPOCH, planned_epochs=NUM_EPOCHS),
                summary_key=SUMMARY_KEY, skip_empty_summaries=SKIP_EMPTY_SUMMARIES,
                selection_metric='rougeL_f1', decoding='greedy', loss='encoder_decoder_summary_token_mean')

def save_summary_checkpoint(tag, model, optimizer, epoch, extra, domain):
    folder = WORK_DIR / 'checkpoints' / tag
    folder.mkdir(parents=True, exist_ok=True)
    rng = {'python': random.getstate(), 'torch': torch.get_rng_state()}
    if torch.cuda.is_available(): rng['cuda'] = torch.cuda.get_rng_state_all()
    payload = dict(model=model.state_dict(), optimizer=optimizer.state_dict(), rng=rng,
                   epoch=epoch, extra=extra, metadata=experiment_metadata(domain))
    tmp = folder / 'checkpoint.tmp'
    torch.save(payload, tmp)
    os.replace(tmp, folder / 'checkpoint.pt')
    if UPLOAD_CHECKPOINTS:
        fs.put_file(str(folder / 'checkpoint.pt'), checkpoint_remote_path(tag))

def load_summary_checkpoint(tag, model, domain, optimizer=None, restore_rng=False):
    path = WORK_DIR / 'checkpoints' / tag / 'checkpoint.pt'
    if not path.exists() or restore_rng:
        if UPLOAD_CHECKPOINTS:
            path = download_bucket_file(checkpoint_remote_path(tag))
        elif not path.exists():
            raise FileNotFoundError(path)
    payload = torch.load(path, map_location='cpu', weights_only=False)  # own trusted checkpoints only
    assert payload['metadata'] == experiment_metadata(domain), 'Checkpoint configuration mismatch'
    model.load_state_dict(payload['model'])
    if optimizer is not None: optimizer.load_state_dict(payload['optimizer'])
    if restore_rng:
        random.setstate(payload['rng']['python'])
        torch.set_rng_state(payload['rng']['torch'])
        if torch.cuda.is_available() and 'cuda' in payload['rng']:
            torch.cuda.set_rng_state_all(payload['rng']['cuda'])
    return payload


class ValidationLRScheduler:
    def __init__(self, optimizer, total_updates):
        assert total_updates > 0 and 0 < WARMUP_RATIO < 1
        assert 0 < WARMUP_START_FACTOR <= 1 and 0 < LR_FACTOR < 1
        assert LR_BAD_CHECKS >= 1 and 0 <= LR_MIN <= LR
        self.optimizer = optimizer
        self.updates = 0
        self.warmup_updates = max(1, math.ceil(total_updates * WARMUP_RATIO))
        self.peak_lrs = [g['lr'] for g in optimizer.param_groups]
        # PyTorch reduces after num_bad_epochs > patience: 1 means two bad checks.
        self.plateau = torch.optim.lr_scheduler.ReduceLROnPlateau(
            optimizer, mode='min', factor=LR_FACTOR, patience=LR_BAD_CHECKS-1,
            threshold=LR_LOSS_THRESHOLD, threshold_mode='rel',
            cooldown=LR_COOLDOWN_CHECKS, min_lr=LR_MIN)
        self._set_warmup_lr()

    def _set_warmup_lr(self):
        factor = WARMUP_START_FACTOR + (1-WARMUP_START_FACTOR) * self.updates/self.warmup_updates
        for group, peak in zip(self.optimizer.param_groups, self.peak_lrs):
            group['lr'] = peak * factor

    def optimizer_step_completed(self):
        self.updates += 1
        if self.updates <= self.warmup_updates: self._set_warmup_lr()

    def validation_step(self, loss):
        if not math.isfinite(loss): raise ValueError('Non-finite validation loss')
        # Warmup and plateau never fight over the learning rate.
        if self.updates >= self.warmup_updates: self.plateau.step(loss)

    def state_dict(self):
        return dict(updates=self.updates, warmup_updates=self.warmup_updates,
                    peak_lrs=self.peak_lrs, plateau=self.plateau.state_dict(),
                    current_lrs=[g['lr'] for g in self.optimizer.param_groups])

    def load_state_dict(self, state):
        if state['warmup_updates'] != self.warmup_updates:
            raise ValueError('Warmup horizon changed; resume with the original dataset, epochs and accumulation')
        self.updates = state['updates']
        self.peak_lrs = state['peak_lrs']
        self.plateau.load_state_dict(state['plateau'])
        for group, lr in zip(self.optimizer.param_groups, state['current_lrs']): group['lr'] = lr

@torch.no_grad()
def evaluate_summary_loss(model, loader):
    was_training = model.training
    model.eval()
    try:
        total_loss = total_tokens = 0
        for batch in loader:
            loss, count = summary_loss_sum(model, batch)
            total_loss += loss.item(); total_tokens += count
        if total_tokens == 0: raise ValueError('No validation summary targets')
        return total_loss/total_tokens
    finally:
        model.train(was_training)

@torch.no_grad()
def evaluate_summary(model, loader):
    model.eval()
    validation_loss = evaluate_summary_loss(model, loader)
    scorer = rouge_scorer.RougeScorer(['rouge1','rougeL'],use_stemmer=True)
    records = []
    started = time.perf_counter()
    with merged_lorma_for_generation(model):
        for row in tqdm(datasets['dev'].rows,desc='Generate validation summaries'):
            prediction, tokens = generate_summary(model,row['chunks'])
            scores = scorer.score(row['reference'],prediction)
            words = prediction.lower().split()
            trigrams = list(zip(words,words[1:],words[2:]))
            records.append(dict(id=row['doc_id'],summary=prediction,reference=row['reference'],
                                rouge1_f1=scores['rouge1'].fmeasure,rougeL_f1=scores['rougeL'].fmeasure,
                                generated_tokens=tokens,repeated_trigram_fraction=1-len(set(trigrams))/len(trigrams) if trigrams else 0.0))
    metrics = {key:sum(r[key] for r in records)/len(records) for key in ['rouge1_f1','rougeL_f1','generated_tokens','repeated_trigram_fraction']}
    metrics.update(loss=validation_loss, generation_seconds=time.perf_counter()-started)
    with open(WORK_DIR / 'SUMM_dev_predictions.json','w',encoding='utf-8') as f:
        json.dump(records,f,ensure_ascii=False,indent=2)
    return metrics

def cleanup_latest_checkpoint():
    # Exact file deletion only. Never delete a parent directory or a Phase-1 file.
    best_local = WORK_DIR / 'checkpoints' / BEST_TAG / 'checkpoint.pt'
    if not best_local.is_file(): raise RuntimeError('Best checkpoint is missing; keeping latest')
    if UPLOAD_CHECKPOINTS:
        best_remote = checkpoint_remote_path(BEST_TAG)
        if not fs.exists(best_remote) or fs.info(best_remote)['size'] != best_local.stat().st_size:
            raise RuntimeError('Best upload not verified; keeping latest')
        latest_remote = checkpoint_remote_path(SAVE_TAG)
        assert latest_remote != best_remote and latest_remote != bucket_path(MODEL_PATH)
        if fs.exists(latest_remote): fs.rm(latest_remote)
        if fs.exists(latest_remote): raise RuntimeError('Latest bucket checkpoint deletion failed')
    latest_local = WORK_DIR / 'checkpoints' / SAVE_TAG / 'checkpoint.pt'
    if latest_local.exists(): latest_local.unlink()
    print('Training completed: retained best checkpoint only:', BEST_TAG)
```

```python
def train_summary(domain='SUMM'):
    assert domain == 'SUMM'
    global TEACHER
    seed_everything(SEED)
    TEACHER = build_teacher()
    assert_run_destination()
    model = build_summary_model()
    check_generation_cache(model)
    check_document_gradients(model)
    check_gradient_replay(model)
    check_distillation_terms(model)
    if sum(p.numel() for p in model.parameters()) >= TEACHER_INFO['parameters']:
        raise ValueError('Student is not smaller than the loaded teacher')
    params = [p for p in model.parameters() if p.requires_grad]
    optimizer = torch.optim.AdamW(params, lr=LR, weight_decay=WEIGHT_DECAY)
    scaler = torch.amp.GradScaler('cuda', enabled=USE_AMP and AMP_DTYPE == torch.float16)
    tag = SAVE_TAG
    best_tag = BEST_TAG
    train_loader, dev_loader = make_loader('train'), make_loader('dev')
    assert len(train_loader) > 0 and VAL_LOSS_CHECKS_PER_EPOCH >= 1
    updates_per_epoch = math.ceil(len(train_loader)/GRAD_ACCUM_STEPS)
    check_interval = max(1, math.ceil(updates_per_epoch/VAL_LOSS_CHECKS_PER_EPOCH))
    scheduler = ValidationLRScheduler(optimizer, NUM_EPOCHS * updates_per_epoch)
    print('Optimizer updates per epoch:', updates_per_epoch, '| warmup updates:', scheduler.warmup_updates)
    start = 0
    extra = dict(best_f1=-1.0, patience_reference=-1.0, stale=0, history=[], best_tag=best_tag)
    if RESUME:
        checkpoint = load_summary_checkpoint(SAVE_TAG, model, domain, optimizer, restore_rng=True)
        start, extra = checkpoint['epoch'], checkpoint['extra']
        best_tag = extra['best_tag']
        if 'scaler' in extra: scaler.load_state_dict(extra['scaler'])
        if 'scheduler' not in extra: raise ValueError('This checkpoint predates the scheduler; start a new run tag')
        scheduler.load_state_dict(extra['scheduler'])
        del checkpoint
    for epoch in range(start, NUM_EPOCHS):
        if extra['stale'] >= PATIENCE: break
        model.train()
        optimizer.zero_grad(set_to_none=True)
        train_sum = train_n = group_n = seen_docs = 0
        next_log = LOG_EVERY_DOCS
        epoch_update = 0
        for step, batch in enumerate(tqdm(train_loader, desc=f'{domain} epoch {epoch+1}')):
            loss_sum, token_count = backward_document_replay(model,batch,scaler)
            group_n += token_count
            train_n += token_count
            train_sum += loss_sum.item()
            seen_docs += len(batch['doc_ids'])
            if seen_docs >= next_log:
                print(f'Epoch {epoch+1} | docs {seen_docs:,}/{len(datasets["train"]):,} | distillation objective {train_sum/train_n:.5f} | CE {model.last_kd_terms["reference_ce"]:.4f} | KL {model.last_kd_terms["teacher_kl"]:.4f}')
                next_log += LOG_EVERY_DOCS
            if (step + 1) % GRAD_ACCUM_STEPS == 0 or step + 1 == len(train_loader):
                scaler.unscale_(optimizer)
                for p in params:
                    if p.grad is not None: p.grad.div_(group_n)
                torch.nn.utils.clip_grad_norm_(params, GRAD_CLIP)
                previous_scale = scaler.get_scale()
                scaler.step(optimizer)
                scaler.update()
                update_succeeded = scaler.get_scale() >= previous_scale
                if update_succeeded: scheduler.optimizer_step_completed()
                optimizer.zero_grad(set_to_none=True)
                group_n = 0
                epoch_update += 1
                if epoch_update % check_interval == 0 or step + 1 == len(train_loader):
                    dev_loss = evaluate_summary_loss(model, dev_loader)
                    old_lr = optimizer.param_groups[0]['lr']
                    scheduler.validation_step(dev_loss)
                    current_lr = optimizer.param_groups[0]['lr']
                    event = dict(epoch=epoch+1, optimizer_updates=scheduler.updates,
                                 dev_loss=dev_loss, lr_before=old_lr, lr=current_lr)
                    extra.setdefault('lr_history', []).append(event)
                    print(f"Validation | update {scheduler.updates:,} | summary loss {dev_loss:.5f} | LR {old_lr:.3g} -> {current_lr:.3g}")
        metrics = evaluate_summary(model, dev_loader)
        f1 = metrics['rougeL_f1']
        improved = f1 > extra['best_f1']
        if improved:
            extra['best_f1'] = f1
        if f1 > extra['patience_reference'] + MIN_DELTA:
            extra['patience_reference'], extra['stale'] = f1, 0
        else: extra['stale'] += 1
        extra['history'].append(dict(epoch=epoch+1, train_loss=train_sum/train_n, lr=optimizer.param_groups[0]['lr'], **metrics))
        print(f"SUMM train loss={train_sum/train_n:.4f}, dev ROUGE-L F1={f1:.4f}, ROUGE-1 F1={metrics['rouge1_f1']:.4f}, patience={extra['stale']}/{PATIENCE}")
        extra['scaler'] = scaler.state_dict()
        extra['scheduler'] = scheduler.state_dict()
        if improved: save_summary_checkpoint(best_tag, model, optimizer, epoch+1, extra, domain)
        save_summary_checkpoint(tag, model, optimizer, epoch+1, extra, domain)
    best_payload = load_summary_checkpoint(best_tag, model, domain)
    del best_payload
    metrics = evaluate_summary(model, dev_loader)
    with open(WORK_DIR / f'{domain}_dev_metrics.json', 'w') as f: json.dump(metrics, f, indent=2)
    with open(WORK_DIR / f'{domain}_history.json', 'w') as f: json.dump(extra['history'], f, indent=2)
    with open(WORK_DIR / 'SUMM_lr_history.json', 'w') as f: json.dump(extra.get('lr_history', []), f, indent=2)
    print('Best development metrics:', metrics)
    compare_teacher_student(model,metrics)
    cleanup_latest_checkpoint()
    del optimizer
    TEACHER = None
    gc.collect()
    if torch.cuda.is_available(): torch.cuda.empty_cache()
    return model, extra['history']
```

```python
def validate_predictions(predictions, documents):
    assert len(predictions) == len(documents)
    assert [p['id'] for p in predictions] == [d['id'] for d in documents]
    assert len({p['id'] for p in predictions}) == len(predictions)
    assert all(set(p) == {'id','summary'} and isinstance(p['summary'],str) for p in predictions)

@torch.no_grad()
def predict_summary(model):
    model.eval()
    predictions = []
    with merged_lorma_for_generation(model):
        for row in tqdm(datasets['test'].rows,desc='SUMM test predictions'):
            summary, _ = generate_summary(model,row['chunks'])
            predictions.append(dict(id=row['doc_id'],summary=summary))
    validate_predictions(predictions,data['test'])
    return predictions

summ_model, summ_history = train_summary()
# Load test only after training; test text is never used for fitting/selection.
data['test'] = normalize_documents(download_json(f'{DATA_PREFIX}/Test/test.json'),labelled=False)
datasets['test'] = SummaryDataset(data['test'],'test')
predictions = predict_summary(summ_model)
json_path = WORK_DIR / 'prediction.json'
with open(json_path,'w',encoding='utf-8') as f:
    json.dump(predictions,f,ensure_ascii=False,indent=2)
del summ_model
gc.collect()
if torch.cuda.is_available(): torch.cuda.empty_cache()
```
