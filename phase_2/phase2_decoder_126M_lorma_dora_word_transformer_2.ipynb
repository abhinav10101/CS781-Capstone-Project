# LoRMA + DoRA with added bidirectional word Transformer layers
Original pretrained GQA decoder (8 causal blocks) → original subword-to-word mean pooling → 1 new bidirectional word Transformer block → original linear BIO classifier. Set EXTRA_WORD_LAYERS=2 for the depth comparison. No external pretrained weights are used.

New blocks use pre-norm LayerNorm, 8-head attention, a 2× width GELU FFN, dropout=0.1 and word-padding masks. Attention and FFN output projections are zero-initialized, preserving the original classifier outputs at initialization. New blocks receive the order-aware pretrained representations; no separate positional embedding is added. The original LoRMA/DoRA adapter and last-two-backbone-block fine-tuning are retained.

Use Restart & Run All from decoder_126M. Old NER checkpoints lack the new blocks. Early stopping retains latest and best checkpoints and restores best weights before prediction. Different depths get separate tags. Validation now uses original gold offsets and a 90/10 document split. Baseline word grouping and non-overlapping chunk decoding are retained, so their boundary/context limitations remain. This is a depth/context experiment, not a character-boundary correction. Class weights and accumulation variables from the original notebook remain unused by its training loop.

CPU tests cover initial identity, word-padding invariance, new-layer gradients, backbone transfer, and NER forward/backward. GPU training and test F1 are unverified.

## Checkpoint related functions

```python
def upload_checkpoint(tmp_dir, tag):

    remote_dir = f"checkpoints/{tag}"

    for filename in os.listdir(tmp_dir):
        local_path = os.path.join(tmp_dir, filename)

        upload_to_bucket(
            local_path,
            f"{remote_dir}/{filename}",
        )
```

```python
def save_checkpoint(tag, model, optimizer, scheduler=None, scaler=None,
                    tokens_seen=0, global_step=0, epoch=0, extra_state=None):

    with tempfile.TemporaryDirectory() as tmp_dir:

        torch.save(model.state_dict(), os.path.join(tmp_dir, "model.pt"))
        torch.save(optimizer.state_dict(), os.path.join(tmp_dir, "optimizer.pt"))

        if scheduler is not None:
            torch.save(scheduler.state_dict(), os.path.join(tmp_dir, "scheduler.pt"))

        if scaler is not None:
            torch.save(scaler.state_dict(), os.path.join(tmp_dir, "scaler.pt"))

        rng_state = {"python": random.getstate(), "torch": torch.get_rng_state()}

        if torch.cuda.is_available():
            rng_state["cuda"] = torch.cuda.get_rng_state_all()

        torch.save(rng_state, os.path.join(tmp_dir, "rng_state.pt"))

        state = {
            "tag": tag,
            "tokens_seen": tokens_seen,
            "global_step": global_step,
            "epoch": epoch,
            "extra_state": extra_state,
        }

        with open(os.path.join(tmp_dir, "training_state.json"), "w") as f:
            json.dump(state, f, indent=2)

        upload_checkpoint(tmp_dir,tag)

def load_checkpoint(tag, model, optimizer, scheduler=None, scaler=None):

    with tempfile.TemporaryDirectory() as tmp_dir:

        files = [
            "model.pt",
            "optimizer.pt",
            "scheduler.pt",
            "scaler.pt",
            "rng_state.pt",
            "training_state.json",
        ]

        checkpoint_dir = os.path.join(
            tmp_dir,
            "checkpoints",
            tag,
        )

        os.makedirs(checkpoint_dir, exist_ok=True)

        for filename in files:
            try:
                download_from_bucket(
                    f"checkpoints/{tag}/{filename}",
                    os.path.join(checkpoint_dir, filename),
                )
            except Exception:
                pass

        model.load_state_dict(
            torch.load(
                os.path.join(checkpoint_dir, "model.pt"),
                map_location=DEVICE,
                weights_only=True,
            )
        )

        optimizer.load_state_dict(
            torch.load(
                os.path.join(checkpoint_dir, "optimizer.pt"),
                map_location=DEVICE,
                weights_only=False,
            )
        )

        scheduler_path = os.path.join(checkpoint_dir, "scheduler.pt")

        if scheduler is not None and os.path.exists(scheduler_path):
            scheduler.load_state_dict(
                torch.load(
                    scheduler_path,
                    map_location=DEVICE,
                    weights_only=False,
                )
            )

        scaler_path = os.path.join(checkpoint_dir, "scaler.pt")

        if scaler is not None and os.path.exists(scaler_path):
            scaler.load_state_dict(
                torch.load(
                    scaler_path,
                    map_location=DEVICE,
                    weights_only=False,
                )
            )

        rng_state = torch.load(
            os.path.join(checkpoint_dir, "rng_state.pt"),
            map_location="cpu",
            weights_only=False,
        )

        random.setstate(rng_state["python"])
        torch.set_rng_state(rng_state["torch"])

        if torch.cuda.is_available() and "cuda" in rng_state:
            torch.cuda.set_rng_state_all(rng_state["cuda"])

        with open(
            os.path.join(checkpoint_dir, "training_state.json")
        ) as f:
            state = json.load(f)

    return state
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


# ============================================================
# RoPE
# ============================================================

class RotaryEmbedding(nn.Module):
    def __init__(self, head_dim, max_seq_len=512, base=10000):
        super().__init__()

        inv_freq = 1.0 / (
            base ** (torch.arange(0, head_dim, 2).float() / head_dim)
        )

        positions = torch.arange(max_seq_len).float()
        freqs = torch.outer(positions, inv_freq)

        self.register_buffer("cos", freqs.cos(), persistent=False)
        self.register_buffer("sin", freqs.sin(), persistent=False)

    def forward(self, q, k):
        # q, k: [B, H, T, D]
        T = q.size(2)

        cos = self.cos[:T].unsqueeze(0).unsqueeze(0)
        sin = self.sin[:T].unsqueeze(0).unsqueeze(0)

        q1, q2 = q[..., ::2], q[..., 1::2]
        k1, k2 = k[..., ::2], k[..., 1::2]

        q_rot = torch.stack(
            [q1 * cos - q2 * sin,
             q1 * sin + q2 * cos],
            dim=-1,
        ).flatten(-2)

        k_rot = torch.stack(
            [k1 * cos - k2 * sin,
             k1 * sin + k2 * cos],
            dim=-1,
        ).flatten(-2)

        return q_rot, k_rot


# ============================================================
# SwiGLU
# ============================================================

class SwiGLU(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()

        self.gate_proj = nn.Linear(d_model, d_ff, bias=False)
        self.up_proj = nn.Linear(d_model, d_ff, bias=False)
        self.down_proj = nn.Linear(d_ff, d_model, bias=False)

    def forward(self, x):
        return self.down_proj(
            F.silu(self.gate_proj(x)) * self.up_proj(x)
        )


# ============================================================
# GQA + SDPA
# ============================================================

class GQAAttention(nn.Module):
    def __init__(self, d_model, n_heads, n_kv_heads):
        super().__init__()

        assert d_model % n_heads == 0
        assert n_heads % n_kv_heads == 0

        self.n_heads = n_heads
        self.n_kv_heads = n_kv_heads
        self.head_dim = d_model // n_heads

        self.q_proj = nn.Linear(
            d_model,
            n_heads * self.head_dim,
            bias=False,
        )

        self.k_proj = nn.Linear(
            d_model,
            n_kv_heads * self.head_dim,
            bias=False,
        )

        self.v_proj = nn.Linear(
            d_model,
            n_kv_heads * self.head_dim,
            bias=False,
        )

        self.out_proj = nn.Linear(
            d_model,
            d_model,
            bias=False,
        )

        self.rope = RotaryEmbedding(
            self.head_dim,
            max_seq_len=512,
        )

    def forward(self, x):

        B, T, _ = x.shape

        q = self.q_proj(x)
        k = self.k_proj(x)
        v = self.v_proj(x)

        q = q.view(
            B, T,
            self.n_heads,
            self.head_dim
        ).transpose(1, 2)

        k = k.view(
            B, T,
            self.n_kv_heads,
            self.head_dim
        ).transpose(1, 2)

        v = v.view(
            B, T,
            self.n_kv_heads,
            self.head_dim
        ).transpose(1, 2)

        q, k = self.rope(q, k)

        # Repeat KV heads for GQA
        repeat_factor = self.n_heads // self.n_kv_heads

        k = k.repeat_interleave(
            repeat_factor,
            dim=1,
        )

        v = v.repeat_interleave(
            repeat_factor,
            dim=1,
        )

        # PyTorch SDPA can dispatch to Flash Attention
        # / memory-efficient kernels automatically.
        attn_out = F.scaled_dot_product_attention(
            q,
            k,
            v,
            is_causal=True,
        )

        attn_out = (
            attn_out
            .transpose(1, 2)
            .contiguous()
            .view(B, T, -1)
        )

        return self.out_proj(attn_out)


# ============================================================
# Decoder Block
# ============================================================

class DecoderBlock(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads,
        n_kv_heads,
        d_ff,
    ):
        super().__init__()

        self.norm1 = nn.RMSNorm(d_model)
        self.attention = GQAAttention(
            d_model,
            n_heads,
            n_kv_heads,
        )

        self.norm2 = nn.RMSNorm(d_model)
        self.ffn = SwiGLU(
            d_model,
            d_ff,
        )

    def forward(self, x):

        x = x + self.attention(
            self.norm1(x)
        )

        x = x + self.ffn(
            self.norm2(x)
        )

        return x


# ============================================================
# Decoder-only Language Model
# ============================================================

class EfficientDecoderLM(nn.Module):

    def __init__(
        self,
        vocab_size,
        d_model=768,
        n_heads=12,
        n_kv_heads=4,
        n_layers=12,
        d_ff=3072,
        max_seq_len=512,
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.d_model = d_model

        self.token_embedding = nn.Embedding(                             
            vocab_size,
            d_model,
        )

        self.blocks = nn.ModuleList([
            DecoderBlock(
                d_model=d_model,
                n_heads=n_heads,
                n_kv_heads=n_kv_heads,
                d_ff=d_ff,
            )
            for _ in range(n_layers)
        ])

        self.final_norm = nn.RMSNorm(d_model)

        # Tied LM head
        self.lm_head = nn.Linear(
            d_model,
            vocab_size,
            bias=False,
        )

        self.lm_head.weight = self.token_embedding.weight

        self.apply(self._init_weights)

    def _init_weights(self, module):

        if isinstance(module, nn.Linear):
            nn.init.xavier_uniform_(module.weight)

        elif isinstance(module, nn.Embedding):
            nn.init.xavier_uniform_(module.weight)

        elif isinstance(module, nn.RMSNorm):
            nn.init.ones_(module.weight)

    def forward(self, input_ids):

        x = self.token_embedding(input_ids)

        for block in self.blocks:
            x = block(x)

        x = self.final_norm(x)

        return self.lm_head(x)
import math
import torch
import torch.nn as nn
import torch.nn.functional as F


class LormaDoRALinear(nn.Module):
    def __init__(
        self,
        original_linear,
        r=16,
        alpha=32,
        dropout=0.05,
        do_pi=True,
        eps=1e-8,
    ):
        super().__init__()

        self.in_features = original_linear.in_features
        self.out_features = original_linear.out_features
        self.r = r
        self.alpha = alpha
        self.scaling = alpha / r
        self.do_pi = do_pi
        self.eps = eps

        self.dropout = nn.Dropout(dropout)

        # Preserve the original pretrained linear layer.
        self.original_linear = original_linear

        # Freeze pretrained weights and bias.
        for param in self.original_linear.parameters():
            param.requires_grad = False

        device = original_linear.weight.device
        dtype = original_linear.weight.dtype

        # LoRMA pre-transformation:
        # B: [out_features, r]
        # A: [r, out_features]
        self.B = nn.Parameter(
            torch.empty(
                self.out_features, r,
                device=device, dtype=dtype,
            )
        )
        self.A = nn.Parameter(
            torch.empty(
                r, self.out_features,
                device=device, dtype=dtype,
            )
        )

        # Initialize the transformation near identity.
        # BA = I when r >= out_features and the initialization
        # is constructed as below; otherwise it is low rank.
        nn.init.zeros_(self.B)
        nn.init.zeros_(self.A)

        with torch.no_grad():
            # This construction makes B @ A = I for any r >= 1:
            # B[:, 0] = 1, A[0, :] = 1 does NOT produce identity.
            # Instead, use a rectangular identity factorization
            # when rank is sufficient.
            k = min(r, self.out_features)
            self.B[:k, :k] = torch.eye(
                k, device=device, dtype=dtype
            )
            self.A[:k, :k] = torch.eye(
                k, device=device, dtype=dtype
            )

        # DoRA-style row magnitudes, initialized from pretrained W.
        initial_magnitude = torch.linalg.vector_norm(
            original_linear.weight.detach().float(),
            ord=2,
            dim=1,
        ).to(device=device, dtype=dtype)

        self.magnitude = nn.Parameter(initial_magnitude)

    def _shift(self, matrix):
        """
        Row-dependent cyclic column shift.
        Matches the shift convention used in our LoRMA implementation.
        """
        n_rows, n_cols = matrix.shape

        row_indices = torch.arange(
            n_rows, device=matrix.device
        ).unsqueeze(1)

        col_indices = torch.arange(
            n_cols, device=matrix.device
        ).unsqueeze(0)

        indices = (col_indices - row_indices) % n_cols

        return torch.gather(matrix, 1, indices)

    def effective_weight(self):
        W = self.original_linear.weight
    
        P_delta = self.B @ self.A
    
        if self.do_pi:
            P_delta = self._shift(P_delta)
    
        identity = torch.eye(
            self.out_features,
            device=W.device,
            dtype=W.dtype,
        )
    
        P = identity + self.scaling * P_delta
        transformed_W = P @ W
    
        row_norms = torch.linalg.vector_norm(
            transformed_W.float(),
            ord=2,
            dim=1,
            keepdim=True,
        ).to(dtype=transformed_W.dtype)
    
        normalized_W = transformed_W / row_norms.clamp_min(self.eps)
    
        return self.magnitude.unsqueeze(1) * normalized_W

    def forward(self, x):
        W_eff = self.effective_weight()

        return F.linear(
            self.dropout(x),
            W_eff,
            self.original_linear.bias,
        )

class BidirectionalWordBlock(nn.Module):
    """Pre-norm word Transformer, initialized to an exact residual identity."""
    def __init__(self,d_model,n_heads=8,d_ff=None,dropout=0.1):
        super().__init__()
        self.norm1=nn.LayerNorm(d_model)
        self.attention=nn.MultiheadAttention(d_model,n_heads,dropout=dropout,batch_first=True)
        self.norm2=nn.LayerNorm(d_model)
        self.ffn=nn.Sequential(nn.Linear(d_model,d_ff or d_model*2),nn.GELU(),
                               nn.Dropout(dropout),nn.Linear(d_ff or d_model*2,d_model))
        self.dropout=nn.Dropout(dropout)
        nn.init.zeros_(self.attention.out_proj.weight)
        nn.init.zeros_(self.attention.out_proj.bias)
        nn.init.zeros_(self.ffn[-1].weight)
        nn.init.zeros_(self.ffn[-1].bias)
    def forward(self,x,padding_mask):
        h=self.norm1(x)
        h,_=self.attention(h,h,h,key_padding_mask=padding_mask,need_weights=False)
        x=x+self.dropout(h)
        x=x+self.dropout(self.ffn(self.norm2(x)))
        return x.masked_fill(padding_mask[:,:,None],0)

class EfficientDecoderLMForNER(EfficientDecoderLM):

    def __init__(
        self,
        vocab_size,
        num_labels=25,
        d_model=768,
        n_heads=12,
        n_kv_heads=4,
        n_layers=12,
        d_ff=3072,
        max_seq_len=512,
        extra_word_layers=1,
        word_dropout=0.1,
    ):
        super().__init__(
            vocab_size=vocab_size,
            d_model=d_model,
            n_heads=n_heads,
            n_kv_heads=n_kv_heads,
            n_layers=n_layers,
            d_ff=d_ff,
            max_seq_len=max_seq_len,
        )

        self.ner_head = nn.Linear(d_model, num_labels)
        self.word_blocks=nn.ModuleList([
            BidirectionalWordBlock(d_model,n_heads=n_heads,dropout=word_dropout)
            for _ in range(extra_word_layers)
        ])

    def forward(self, input_ids, word_groups):
    # input_ids: [B, T]
        B = input_ids.size(0)
    
        x = self.token_embedding(input_ids)
    
        for block in self.blocks:
            x = block(x)
    
        x = self.final_norm(x)
    
        batch_word_hidden = []
        max_words = 0
    
        for b in range(B):
    
            word_hidden = []
    
            for group in word_groups[b]:
                h = x[b, group].mean(dim=0)
                word_hidden.append(h)
    
            word_hidden = torch.stack(word_hidden)
    
            batch_word_hidden.append(word_hidden)
    
            max_words = max(
                max_words,
                word_hidden.size(0)
            )
    
        # [B, max_words, D]
        padded_word_hidden = x.new_zeros(
            B,
            max_words,
            x.size(-1)
        )
    
        for b, word_hidden in enumerate(batch_word_hidden):
    
            n_words = word_hidden.size(0)
    
            padded_word_hidden[
                b, :n_words
            ] = word_hidden
    
        # Zero-initialized residual outputs preserve the original head at initialization.
        lengths=torch.tensor([len(g) for g in word_groups],device=x.device)
        padding_mask=torch.arange(max_words,device=x.device)[None,:]>=lengths[:,None]
        # The pretrained causal representations already encode position/order.
        for block in self.word_blocks:
            padded_word_hidden=block(padded_word_hidden,padding_mask)

        # [B, max_words, num_labels]
        logits = self.ner_head(padded_word_hidden)
    
        return logits
```

```python
def prepare_lner_data(text_data, gold_data, tokenizer):

    processed_data = {}

    for doc_id, text in text_data.items():

       

        # --------------------------------------------------
        # 1. Tokenize
        # --------------------------------------------------

        encoding = tokenizer.encode(text)

        tokens = encoding.tokens
        input_ids = encoding.ids
        offsets = encoding.offsets

        # --------------------------------------------------
        # 2. Group tokens
        #
        # Rule:
        #   Ġ  -> new word
        #   no Ġ -> continue previous word
        # --------------------------------------------------

        word_groups = []
        current_group = []

        for i, token in enumerate(tokens):

            start, end = offsets[i]

            if start == end:
                continue

            if current_group and token.startswith("Ġ"):
                word_groups.append(current_group)
                current_group = []

            current_group.append(i)

        if current_group:
            word_groups.append(current_group)

        # --------------------------------------------------
        # 3. Correct character spans
        #
        # Raw tokenizer offsets include the whitespace
        # represented by Ġ.
        # Remove that whitespace from the beginning.
        # --------------------------------------------------

        word_spans = []

        for group in word_groups:

            start = offsets[group[0]][0]
            end = offsets[group[-1]][1]

            while start < end and text[start].isspace():
                start += 1

            word_spans.append((start, end))

        # --------------------------------------------------
        # 4. Initialize labels
        # --------------------------------------------------

        word_labels = ["O"] * len(word_groups)

        # --------------------------------------------------
        # 5. Align gold entities
        # --------------------------------------------------
        if gold_data is not None:
            gold_entities = gold_data[doc_id]
            for entity_start, entity_end, entity_type, entity_text in gold_entities:
    
                overlapping_words = []
    
                for word_idx, (word_start, word_end) in enumerate(word_spans):
    
                    if word_start < entity_end and word_end > entity_start:
                        overlapping_words.append(word_idx)
    
                if not overlapping_words:
                    continue
    
                word_labels[overlapping_words[0]] = f"B-{entity_type}"
    
                for word_idx in overlapping_words[1:]:
                    word_labels[word_idx] = f"I-{entity_type}"

        gold_spans=[]
        if gold_data is not None:
            for a,b,label,entity_text in gold_data[doc_id]:
                if not(0<=a<b<=len(text)) or text[a:b]!=entity_text:
                    raise ValueError(f'Gold offset mismatch: {doc_id}, {(a,b)}')
                gold_spans.append([a,b,label])
        # --------------------------------------------------
        # 6. Save
        # --------------------------------------------------

        processed_data[doc_id] = {
            "input_ids": input_ids,
            "offsets": offsets,
            "word_groups": word_groups,
            "word_spans": word_spans,
            "word_labels": word_labels,
            "gold_spans": gold_spans,
        }

    return processed_data
```

```python
MAX_SEQ_LEN = 512

def make_chunks(example, max_seq_len=512):
    """
    Split one processed document into chunks at word boundaries.

    Returns list of:
        {
            "input_ids": [...],
            "word_groups": [...],
            "word_labels": [...]
        }
    """

    input_ids = example["input_ids"]
    word_groups = example["word_groups"]
    word_labels = example["word_labels"]
    word_spans = example["word_spans"]
    chunks = []

    current_words = []
    current_start = None
    current_end = None

    for word_idx, group in enumerate(word_groups):

        word_start = group[0]
        word_end = group[-1] + 1

        if current_start is None:
            current_start = word_start
            current_end = word_end
            current_words = [word_idx]
            continue

        # Would adding this word exceed 2048?
        new_end = word_end

        if new_end - current_start > max_seq_len:

            # Save current chunk
            chunk_input_ids = input_ids[current_start:current_end]

            chunk_groups = [
                [
                    token_idx - current_start
                    for token_idx in word_groups[w]
                ]
                for w in current_words
            ]

            chunk_labels = [
                word_labels[w]
                for w in current_words
            ]
            chunk_spans = [
                word_spans[w]
                for w in current_words
            ]
            chunks.append({
                "input_ids": chunk_input_ids,
                "word_groups": chunk_groups,
                "word_labels": chunk_labels,
                "word_spans" : chunk_spans
            })

            # Start new chunk
            current_start = word_start
            current_end = word_end
            current_words = [word_idx]

        else:
            current_end = word_end
            current_words.append(word_idx)

    # Last chunk
    if current_words:

        chunk_input_ids = input_ids[current_start:current_end]

        chunk_groups = [
            [
                token_idx - current_start
                for token_idx in word_groups[w]
            ]
            for w in current_words
        ]

        chunk_labels = [
            word_labels[w]
            for w in current_words
        ]
        chunk_spans = [
            word_spans[w]
            for w in current_words
        ]
        chunks.append({
            "input_ids": chunk_input_ids,
            "word_groups": chunk_groups,
            "word_labels": chunk_labels,
            "word_spans" : chunk_spans
        })

    return chunks
```

```python
def collate_lner(batch, pad_token_id):

    max_len = max(
        len(x["input_ids"])
        for x in batch
    )

    max_words = max(
        len(x["word_labels"])
        for x in batch
    )

    input_ids = []
    word_groups = []
    word_labels = []

    for x in batch:

        # -------------------------
        # Token padding
        # -------------------------

        ids = x["input_ids"]

        padded = ids + [pad_token_id] * (
            max_len - len(ids)
        )

        input_ids.append(padded)

        # -------------------------
        # Word groups
        # -------------------------

        word_groups.append(
            x["word_groups"]
        )

        # -------------------------
        # Word-label padding
        # -------------------------

        labels = [
            label2id[label]
            for label in x["word_labels"]
        ]

        labels += [-100] * (
            max_words - len(labels)
        )

        word_labels.append(labels)

    return {
        "input_ids": torch.tensor(
            input_ids,
            dtype=torch.long
        ),

        "word_groups": word_groups,

        "word_labels": torch.tensor(
            word_labels,
            dtype=torch.long
        ),
    }
```

```python
def decode_bio_predictions(
    predictions,
    word_spans,
):
    entities = []

    current_label = None
    current_start = None
    current_end = None

    for label, (start, end) in zip(
        predictions,
        word_spans
    ):

        # O
        if label == "O":

            if current_label is not None:
                entities.append([
                    current_start,
                    current_end,
                    current_label
                ])

                current_label = None
                current_start = None
                current_end = None

            continue

        # B-ENTITY
        if label.startswith("B-"):

            if current_label is not None:
                entities.append([
                    current_start,
                    current_end,
                    current_label
                ])

            current_label = label[2:]
            current_start = start
            current_end = end

            continue

        # I-ENTITY
        if label.startswith("I-"):

            entity_type = label[2:]

            # Continue current entity
            if current_label == entity_type:

                current_end = end

            # Invalid I-ENTITY -> treat as a new entity
            else:

                if current_label is not None:
                    entities.append([
                        current_start,
                        current_end,
                        current_label
                    ])

                current_label = entity_type
                current_start = start
                current_end = end

    # Flush final entity
    if current_label is not None:
        entities.append([
            current_start,
            current_end,
            current_label
        ])

    return entities
```

```python
def bio_to_spans(labels, word_spans):
    """
    Convert BIO word labels + character spans into:
        [start, end, label]
    """

    entities = []

    current_label = None
    current_start = None
    current_end = None

    for label, (start, end) in zip(
        labels,
        word_spans
    ):

        if label == "O":

            if current_label is not None:
                entities.append([
                    current_start,
                    current_end,
                    current_label
                ])

                current_label = None
                current_start = None
                current_end = None

            continue

        if label.startswith("B-"):

            if current_label is not None:
                entities.append([
                    current_start,
                    current_end,
                    current_label
                ])

            current_label = label[2:]
            current_start = start
            current_end = end

        elif label.startswith("I-"):

            entity_type = label[2:]

            if current_label == entity_type:

                current_end = end

            else:
                # Treat invalid I-X as a new entity
                if current_label is not None:
                    entities.append([
                        current_start,
                        current_end,
                        current_label
                    ])

                current_label = entity_type
                current_start = start
                current_end = end

    if current_label is not None:
        entities.append([
            current_start,
            current_end,
            current_label
        ])

    return entities

from tqdm import tqdm
import torch


def generate_val_predictions(
    teacher,
    all_processed_data,
    val_doc_ids,
    id2label,
    max_seq_len=512,
):
    teacher.eval()

    predicted = {}
    gold = {}

    with torch.no_grad():

        for doc_id in tqdm(
            val_doc_ids,
            desc="Validation prediction"
        ):

            example = all_processed_data[doc_id]

            chunks = make_chunks(
                example,
                max_seq_len
            )

            doc_predictions = []

            for chunk in chunks:

                input_ids = torch.tensor(
                    [chunk["input_ids"]],
                    dtype=torch.long,
                    device=DEVICE
                )

                word_groups = [
                    chunk["word_groups"]
                ]

                logits = teacher(
                    input_ids,
                    word_groups
                )

                predicted_ids = logits.argmax(
                    dim=-1
                )[0]

                predicted_labels = [
                    id2label[idx.item()]
                    for idx in predicted_ids
                ]

                entities = decode_bio_predictions(
                    predicted_labels,
                    chunk["word_spans"]
                )

                doc_predictions.extend(
                    entities
                )

            predicted[doc_id] = doc_predictions

            gold[doc_id] = example['gold_spans']

    return predicted, gold

from collections import defaultdict


def evaluate_strict_ner(
    predicted,
    gold,
    labels,
):
    """
    Strict entity-level evaluation.

    Returns:
        Macro-F1
        Macro-Precision
        Macro-Recall
        per-label statistics
    """

    stats = {
        label: {
            "tp": 0,
            "fp": 0,
            "fn": 0,
        }
        for label in labels
    }

    for doc_id in gold:

        gold_entities = gold.get(
            doc_id,
            []
        )

        pred_entities = predicted.get(
            doc_id,
            []
        )

        # Exact entity tuples
        gold_set = set(
            tuple(entity)
            for entity in gold_entities
        )

        pred_set = set(
            tuple(entity)
            for entity in pred_entities
        )

        # True positives
        for entity in pred_set & gold_set:

            label = entity[2]

            if label in stats:
                stats[label]["tp"] += 1

        # False positives
        for entity in pred_set - gold_set:

            label = entity[2]

            if label in stats:
                stats[label]["fp"] += 1

        # False negatives
        for entity in gold_set - pred_set:

            label = entity[2]

            if label in stats:
                stats[label]["fn"] += 1

    # -------------------------
    # Per-label metrics
    # -------------------------

    per_label = {}

    macro_precision = 0.0
    macro_recall = 0.0
    macro_f1 = 0.0

    for label in labels:

        tp = stats[label]["tp"]
        fp = stats[label]["fp"]
        fn = stats[label]["fn"]

        precision = (
            tp / (tp + fp)
            if tp + fp > 0
            else 0.0
        )

        recall = (
            tp / (tp + fn)
            if tp + fn > 0
            else 0.0
        )

        f1 = (
            2 * precision * recall /
            (precision + recall)
            if precision + recall > 0
            else 0.0
        )

        per_label[label] = {
            "precision": precision,
            "recall": recall,
            "f1": f1,
            "tp": tp,
            "fp": fp,
            "fn": fn,
        }

        macro_precision += precision
        macro_recall += recall
        macro_f1 += f1

    n = len(labels)

    macro_precision /= n
    macro_recall /= n
    macro_f1 /= n

    return {
        "strict_macro_f1": macro_f1,
        "strict_macro_precision": macro_precision,
        "strict_macro_recall": macro_recall,
        "per_label": per_label,
    }
```

```python
def train_teacher(
    teacher,
    train_loader,
    val_loader,
    label2id,
    optimizer,
    trainable_params,
    tag,
    num_epochs=5,
    resume=True,
    resume_tag=None,
    patience=10,
    min_delta=0.001,
):
    device = next(teacher.parameters()).device

    criterion = nn.CrossEntropyLoss(
        ignore_index=-100
    )

    # =========================
    # RESUME
    # =========================

    if patience < 1 or min_delta < 0:
        raise ValueError("patience must be >= 1 and min_delta must be >= 0")
    start_epoch = 0
    best_f1 = float("-inf")
    patience_reference = float("-inf")
    epochs_without_improvement = 0
    history = []
    best_tag = f"{tag}_best"
    have_best = False

    if resume and resume_tag is not None:

        state = load_checkpoint(
            tag=resume_tag,
            model=teacher,
            optimizer=optimizer,
        )

        start_epoch = state["epoch"]
        extra = state.get("extra_state") or {}
        best_f1 = extra.get("best_f1", float("-inf"))
        patience_reference = extra.get("patience_reference", best_f1)
        epochs_without_improvement = extra.get("epochs_without_improvement", 0)
        history = extra.get("history", [])
        best_tag = extra.get("best_tag", best_tag)
        have_best = extra.get("have_best", False)
        # Older checkpoints have no early-stopping state: tracking starts here.

        print(
            f"Resuming from epoch {start_epoch}"
        )

    # =========================
    # TRAINING
    # =========================

    for epoch in range(start_epoch, num_epochs):
        if epochs_without_improvement >= patience:
            print("Resumed checkpoint already reached early-stopping patience")
            break

        teacher.train()

        total_train_loss = 0.0

        progress = tqdm(
            train_loader,
            desc=f"Epoch {epoch + 1}/{num_epochs}"
        )

        for step, batch in enumerate(progress):

            input_ids = batch["input_ids"].to(device)

            word_groups = batch["word_groups"]

            targets = batch["word_labels"].to(device)

            optimizer.zero_grad(set_to_none=True)

            # [B, max_words, num_labels]
            logits = teacher(
                input_ids,
                word_groups
            )

            # Flatten:
            # logits  -> [B * max_words, num_labels]
            # targets -> [B * max_words]
            loss = criterion(
                logits.view(
                    -1,
                    logits.size(-1)
                ),
                targets.view(-1)
            )

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                trainable_params,
                GRAD_CLIP
            )

            optimizer.step()

            total_train_loss += loss.item()

            progress.set_postfix(
                loss=f"{loss.item():.4f}"
            )

        avg_train_loss = (
            total_train_loss /
            len(train_loader)
        )

        # =========================
        # VALIDATION
        # =========================

        teacher.eval()

        total_val_loss = 0.0
        
        with torch.no_grad():
        
            for batch in val_loader:
        
                input_ids = batch["input_ids"].to(device)
                word_groups = batch["word_groups"]
                targets = batch["word_labels"].to(device)
        
                logits = teacher(
                    input_ids,
                    word_groups
                )
        
                loss = criterion(
                    logits.view(-1, logits.size(-1)),
                    targets.view(-1)
                )
        
                total_val_loss += loss.item()
        
        avg_val_loss = (
            total_val_loss /
            len(val_loader)
        )
        
        # =========================
        # STRICT VALIDATION F1
        # =========================
        
        val_predicted, val_gold = generate_val_predictions(
            teacher=teacher,
            all_processed_data=all_processed_data,
            val_doc_ids=val_doc_ids,
            id2label=id2label,
            max_seq_len=MAX_SEQ_LEN,
        )
        
        val_metrics = evaluate_strict_ner(
            predicted=val_predicted,
            gold=val_gold,
            labels=ENTITY_LABELS,
        )
        
        val_f1 = val_metrics["strict_macro_f1"]
        # =========================
        # LOGGING
        # =========================

        print(
            f"\nEpoch {epoch + 1}/{num_epochs}"
        )

        print(
            f"Train Loss:    {avg_train_loss:.4f}"
        )

        print(
            f"Val Loss:      {avg_val_loss:.4f}"
        )

        print(
            f"Val Strict F1:  {val_f1:.4f}"
        )

        # Save any strict F1 improvement, even if it is smaller than min_delta.
        improved = val_f1 > best_f1
        if improved:
            best_f1 = val_f1
            have_best = True

        # min_delta controls patience resets, not which checkpoint is best.
        if val_f1 > patience_reference + min_delta:
            patience_reference = val_f1
            epochs_without_improvement = 0
        else:
            epochs_without_improvement += 1

        history.append({
            "epoch": epoch + 1,
            "train_loss": avg_train_loss,
            "val_loss": avg_val_loss,
            "val_strict_macro_f1": val_f1,
        })
        extra = {
            "best_f1": best_f1,
            "patience_reference": patience_reference,
            "epochs_without_improvement": epochs_without_improvement,
            "best_tag": best_tag,
            "have_best": have_best,
            "history": history,
        }
        if improved:
            save_checkpoint(
                tag=best_tag, model=teacher, optimizer=optimizer,
                epoch=epoch + 1, extra_state=extra,
            )
            print(f"New best checkpoint: {best_tag} (F1={best_f1:.4f})")

        # Keep latest state separately so an interrupted run can resume.
        save_checkpoint(
            tag=tag, model=teacher, optimizer=optimizer,
            epoch=epoch + 1, extra_state=extra,
        )
        print(f"Early stopping: {epochs_without_improvement}/{patience}")
        if epochs_without_improvement >= patience:
            print(f"Stopping early. Best validation F1: {best_f1:.4f}")
            break

    if have_best:
        # Mutates the existing teacher, so the existing prediction cells use best weights.
        load_checkpoint(tag=best_tag, model=teacher, optimizer=optimizer)
        print(f"Restored best model from {best_tag} (F1={best_f1:.4f})")
    teacher.eval()
    return teacher, history
```

```python
import json
import zipfile
import os
import torch
from tqdm import tqdm


def generate_predictions(
    teacher,
    processed_test,
    id2label,
    tag,
    max_seq_len=512,
):
    teacher.eval()

    predictions = {}

    with torch.no_grad():

        for doc_id, example in tqdm(
            processed_test.items(),
            desc="Generating predictions"
        ):

            chunks = make_chunks(
                example,
                max_seq_len
            )

            document_entities = []

            for chunk in chunks:

                input_ids = torch.tensor(
                    [chunk["input_ids"]],
                    dtype=torch.long,
                    device=DEVICE
                )

                word_groups = [
                    chunk["word_groups"]
                ]

                logits = teacher(
                    input_ids,
                    word_groups
                )

                predicted_ids = logits.argmax(
                    dim=-1
                )[0]

                predicted_labels = [
                    id2label[idx.item()]
                    for idx in predicted_ids
                ]

                entities = decode_bio_predictions(
                    predicted_labels,
                    chunk["word_spans"]
                )

                document_entities.extend(
                    entities
                )

            predictions[doc_id] = document_entities

    # =========================
    # Save predictions.json
    # =========================

    json_path = f"/kaggle/working/predictions.json"

    with open(json_path, "w") as f:
        json.dump(
            predictions,
            f,
            indent=2
        )

    # =========================
    # Zip as tag.zip
    # =========================

    zip_path = f"/kaggle/working/{tag}.zip"

    with zipfile.ZipFile(
        zip_path,
        "w",
        zipfile.ZIP_DEFLATED
    ) as z:

        z.write(
            json_path,
            arcname="prediction.json"
        )

    print(f"Saved: {zip_path}")

    return zip_path
```
