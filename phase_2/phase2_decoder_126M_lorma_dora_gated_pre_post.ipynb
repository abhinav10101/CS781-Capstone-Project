# Gated pre/post LoRMA–DoRA
Each attention projection has separate output-space (pre) and input-space (post) multiplicative branches. A single Linear(in_features, 1) plus sigmoid predicts one gate per token. Both branches share a DoRA magnitude vector. This is a proposed residual shifted multiplicative variant, not a claim of exact official LoRMA implementation.

Restart & Run All. Initialize from `decoder_126M`; old single-branch NER checkpoints are incompatible. Original tokenizer, causal backbone, linear BIO head and last-two-block fine-tuning are retained. New optimizer groups use a lower backbone/adapter LR than head LR. Early stopping saves latest and best checkpoints separately and restores best weights for prediction.

Validation uses original gold offsets and a 90/10 document split. The original word grouping, non-overlapping chunking and BIO decoding remain: some exact character boundaries or overlapping entities cannot be represented by this head. This notebook tests adapter mixing, not a new boundary head. CPU adapter and integration tests do not establish test F1. The old class_weights and GRAD_ACCUM_STEPS cells remain unused by the baseline loop.

## Checkpoint related functions

```python
def upload_checkpoint(tmp_dir, tag):

    remote_dir = f"checkpoints/{tag}"

    for filename in os.listdir(tmp_dir):
        local_path = os.path.join(tmp_dir, filename)

        upload_to_bucket(
            local_path,
            f"{remote_dir}/{filename}",
        )
```

```python
def save_checkpoint(tag, model, optimizer, scheduler=None, scaler=None,
                    tokens_seen=0, global_step=0, epoch=0, extra_state=None):

    with tempfile.TemporaryDirectory() as tmp_dir:

        torch.save(model.state_dict(), os.path.join(tmp_dir, "model.pt"))
        torch.save(optimizer.state_dict(), os.path.join(tmp_dir, "optimizer.pt"))

        if scheduler is not None:
            torch.save(scheduler.state_dict(), os.path.join(tmp_dir, "scheduler.pt"))

        if scaler is not None:
            torch.save(scaler.state_dict(), os.path.join(tmp_dir, "scaler.pt"))

        rng_state = {"python": random.getstate(), "torch": torch.get_rng_state()}

        if torch.cuda.is_available():
            rng_state["cuda"] = torch.cuda.get_rng_state_all()

        torch.save(rng_state, os.path.join(tmp_dir, "rng_state.pt"))

        state = {
            "tag": tag,
            "tokens_seen": tokens_seen,
            "global_step": global_step,
            "epoch": epoch,
            "extra_state": extra_state,
        }

        with open(os.path.join(tmp_dir, "training_state.json"), "w") as f:
            json.dump(state, f, indent=2)

        upload_checkpoint(tmp_dir,tag)

def load_checkpoint(tag, model, optimizer, scheduler=None, scaler=None):

    with tempfile.TemporaryDirectory() as tmp_dir:

        files = [
            "model.pt",
            "optimizer.pt",
            "scheduler.pt",
            "scaler.pt",
            "rng_state.pt",
            "training_state.json",
        ]

        checkpoint_dir = os.path.join(
            tmp_dir,
            "checkpoints",
            tag,
        )

        os.makedirs(checkpoint_dir, exist_ok=True)

        for filename in files:
            try:
                download_from_bucket(
                    f"checkpoints/{tag}/{filename}",
                    os.path.join(checkpoint_dir, filename),
                )
            except Exception:
                pass

        model.load_state_dict(
            torch.load(
                os.path.join(checkpoint_dir, "model.pt"),
                map_location=DEVICE,
                weights_only=True,
            )
        )

        optimizer.load_state_dict(
            torch.load(
                os.path.join(checkpoint_dir, "optimizer.pt"),
                map_location=DEVICE,
                weights_only=False,
            )
        )

        scheduler_path = os.path.join(checkpoint_dir, "scheduler.pt")

        if scheduler is not None and os.path.exists(scheduler_path):
            scheduler.load_state_dict(
                torch.load(
                    scheduler_path,
                    map_location=DEVICE,
                    weights_only=False,
                )
            )

        scaler_path = os.path.join(checkpoint_dir, "scaler.pt")

        if scaler is not None and os.path.exists(scaler_path):
            scaler.load_state_dict(
                torch.load(
                    scaler_path,
                    map_location=DEVICE,
                    weights_only=False,
                )
            )

        rng_state = torch.load(
            os.path.join(checkpoint_dir, "rng_state.pt"),
            map_location="cpu",
            weights_only=False,
        )

        random.setstate(rng_state["python"])
        torch.set_rng_state(rng_state["torch"])

        if torch.cuda.is_available() and "cuda" in rng_state:
            torch.cuda.set_rng_state_all(rng_state["cuda"])

        with open(
            os.path.join(checkpoint_dir, "training_state.json")
        ) as f:
            state = json.load(f)

    return state
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


# ============================================================
# RoPE
# ============================================================

class RotaryEmbedding(nn.Module):
    def __init__(self, head_dim, max_seq_len=512, base=10000):
        super().__init__()

        inv_freq = 1.0 / (
            base ** (torch.arange(0, head_dim, 2).float() / head_dim)
        )

        positions = torch.arange(max_seq_len).float()
        freqs = torch.outer(positions, inv_freq)

        self.register_buffer("cos", freqs.cos(), persistent=False)
        self.register_buffer("sin", freqs.sin(), persistent=False)

    def forward(self, q, k):
        # q, k: [B, H, T, D]
        T = q.size(2)

        cos = self.cos[:T].unsqueeze(0).unsqueeze(0)
        sin = self.sin[:T].unsqueeze(0).unsqueeze(0)

        q1, q2 = q[..., ::2], q[..., 1::2]
        k1, k2 = k[..., ::2], k[..., 1::2]

        q_rot = torch.stack(
            [q1 * cos - q2 * sin,
             q1 * sin + q2 * cos],
            dim=-1,
        ).flatten(-2)

        k_rot = torch.stack(
            [k1 * cos - k2 * sin,
             k1 * sin + k2 * cos],
            dim=-1,
        ).flatten(-2)

        return q_rot, k_rot


# ============================================================
# SwiGLU
# ============================================================

class SwiGLU(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()

        self.gate_proj = nn.Linear(d_model, d_ff, bias=False)
        self.up_proj = nn.Linear(d_model, d_ff, bias=False)
        self.down_proj = nn.Linear(d_ff, d_model, bias=False)

    def forward(self, x):
        return self.down_proj(
            F.silu(self.gate_proj(x)) * self.up_proj(x)
        )


# ============================================================
# GQA + SDPA
# ============================================================

class GQAAttention(nn.Module):
    def __init__(self, d_model, n_heads, n_kv_heads):
        super().__init__()

        assert d_model % n_heads == 0
        assert n_heads % n_kv_heads == 0

        self.n_heads = n_heads
        self.n_kv_heads = n_kv_heads
        self.head_dim = d_model // n_heads

        self.q_proj = nn.Linear(
            d_model,
            n_heads * self.head_dim,
            bias=False,
        )

        self.k_proj = nn.Linear(
            d_model,
            n_kv_heads * self.head_dim,
            bias=False,
        )

        self.v_proj = nn.Linear(
            d_model,
            n_kv_heads * self.head_dim,
            bias=False,
        )

        self.out_proj = nn.Linear(
            d_model,
            d_model,
            bias=False,
        )

        self.rope = RotaryEmbedding(
            self.head_dim,
            max_seq_len=512,
        )

    def forward(self, x):

        B, T, _ = x.shape

        q = self.q_proj(x)
        k = self.k_proj(x)
        v = self.v_proj(x)

        q = q.view(
            B, T,
            self.n_heads,
            self.head_dim
        ).transpose(1, 2)

        k = k.view(
            B, T,
            self.n_kv_heads,
            self.head_dim
        ).transpose(1, 2)

        v = v.view(
            B, T,
            self.n_kv_heads,
            self.head_dim
        ).transpose(1, 2)

        q, k = self.rope(q, k)

        # Repeat KV heads for GQA
        repeat_factor = self.n_heads // self.n_kv_heads

        k = k.repeat_interleave(
            repeat_factor,
            dim=1,
        )

        v = v.repeat_interleave(
            repeat_factor,
            dim=1,
        )

        # PyTorch SDPA can dispatch to Flash Attention
        # / memory-efficient kernels automatically.
        attn_out = F.scaled_dot_product_attention(
            q,
            k,
            v,
            is_causal=True,
        )

        attn_out = (
            attn_out
            .transpose(1, 2)
            .contiguous()
            .view(B, T, -1)
        )

        return self.out_proj(attn_out)


# ============================================================
# Decoder Block
# ============================================================

class DecoderBlock(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads,
        n_kv_heads,
        d_ff,
    ):
        super().__init__()

        self.norm1 = nn.RMSNorm(d_model)
        self.attention = GQAAttention(
            d_model,
            n_heads,
            n_kv_heads,
        )

        self.norm2 = nn.RMSNorm(d_model)
        self.ffn = SwiGLU(
            d_model,
            d_ff,
        )

    def forward(self, x):

        x = x + self.attention(
            self.norm1(x)
        )

        x = x + self.ffn(
            self.norm2(x)
        )

        return x


# ============================================================
# Decoder-only Language Model
# ============================================================

class EfficientDecoderLM(nn.Module):

    def __init__(
        self,
        vocab_size,
        d_model=768,
        n_heads=12,
        n_kv_heads=4,
        n_layers=12,
        d_ff=3072,
        max_seq_len=512,
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.d_model = d_model

        self.token_embedding = nn.Embedding(                             
            vocab_size,
            d_model,
        )

        self.blocks = nn.ModuleList([
            DecoderBlock(
                d_model=d_model,
                n_heads=n_heads,
                n_kv_heads=n_kv_heads,
                d_ff=d_ff,
            )
            for _ in range(n_layers)
        ])

        self.final_norm = nn.RMSNorm(d_model)

        # Tied LM head
        self.lm_head = nn.Linear(
            d_model,
            vocab_size,
            bias=False,
        )

        self.lm_head.weight = self.token_embedding.weight

        self.apply(self._init_weights)

    def _init_weights(self, module):

        if isinstance(module, nn.Linear):
            nn.init.xavier_uniform_(module.weight)

        elif isinstance(module, nn.Embedding):
            nn.init.xavier_uniform_(module.weight)

        elif isinstance(module, nn.RMSNorm):
            nn.init.ones_(module.weight)

    def forward(self, input_ids):

        x = self.token_embedding(input_ids)

        for block in self.blocks:
            x = block(x)

        x = self.final_norm(x)

        return self.lm_head(x)
import math
import torch
import torch.nn as nn
import torch.nn.functional as F


class LormaDoRALinear(nn.Module):
    """Token-dependent mixture of pre/post residual shifted LoRMA + DoRA."""
    def __init__(self,original_linear,r=16,alpha=32,dropout=0.05,do_pi=True,eps=1e-8):
        super().__init__()
        self.original_linear=original_linear
        self.in_features=original_linear.in_features
        self.out_features=original_linear.out_features
        self.scaling=alpha/r;self.do_pi=do_pi;self.eps=eps
        self.dropout=nn.Dropout(dropout)
        for p in original_linear.parameters():p.requires_grad=False
        w=original_linear.weight
        # Pre acts on output coordinates; post acts on input coordinates.
        self.A_pre=nn.Parameter(w.new_empty(r,self.out_features))
        self.B_pre=nn.Parameter(w.new_zeros(self.out_features,r))
        self.A_post=nn.Parameter(w.new_empty(r,self.in_features))
        self.B_post=nn.Parameter(w.new_zeros(self.in_features,r))
        nn.init.normal_(self.A_pre,std=0.02)
        nn.init.normal_(self.A_post,std=0.02)
        self.magnitude=nn.Parameter(w.detach().float().norm(dim=1).to(w.dtype))
        self.gate=nn.Linear(self.in_features,1,device=w.device,dtype=w.dtype)
        nn.init.zeros_(self.gate.weight);nn.init.zeros_(self.gate.bias)
    def _shift(self,matrix):
        n=matrix.size(0)
        idx=(torch.arange(n,device=matrix.device)[None,:]-torch.arange(n,device=matrix.device)[:,None])%n
        return matrix.gather(1,idx)
    def effective_weights(self):
        w=self.original_linear.weight
        left=self.B_pre @ self.A_pre
        right=self.B_post @ self.A_post
        if self.do_pi:left=self._shift(left);right=self._shift(right)
        pre=w+self.scaling*(left @ w)
        post=w+self.scaling*(w @ right)
        def dora(direction):
            norm=direction.float().norm(dim=1,keepdim=True).detach().clamp_min(self.eps)
            return (self.magnitude.float()[:,None]*direction.float()/norm).to(w.dtype)
        return dora(pre),dora(post)
    def forward(self,x):
        w=self.original_linear.weight
        pre,post=self.effective_weights()
        gate=torch.sigmoid(self.gate(x))
        adapted_input=self.dropout(x)
        # Drop only adapter corrections, preserving the pretrained path.
        base=F.linear(x,w,self.original_linear.bias)
        return base+gate*F.linear(adapted_input,pre-w)+(1-gate)*F.linear(adapted_input,post-w)

class EfficientDecoderLMForNER(EfficientDecoderLM):

    def __init__(
        self,
        vocab_size,
        num_labels=25,
        d_model=768,
        n_heads=12,
        n_kv_heads=4,
        n_layers=12,
        d_ff=3072,
        max_seq_len=512,
    ):
        super().__init__(
            vocab_size=vocab_size,
            d_model=d_model,
            n_heads=n_heads,
            n_kv_heads=n_kv_heads,
            n_layers=n_layers,
            d_ff=d_ff,
            max_seq_len=max_seq_len,
        )

        self.ner_head = nn.Linear(d_model, num_labels)

    def forward(self, input_ids, word_groups):
    # input_ids: [B, T]
        B = input_ids.size(0)
    
        x = self.token_embedding(input_ids)
    
        for block in self.blocks:
            x = block(x)
    
        x = self.final_norm(x)
    
        batch_word_hidden = []
        max_words = 0
    
        for b in range(B):
    
            word_hidden = []
    
            for group in word_groups[b]:
                h = x[b, group].mean(dim=0)
                word_hidden.append(h)
    
            word_hidden = torch.stack(word_hidden)
    
            batch_word_hidden.append(word_hidden)
    
            max_words = max(
                max_words,
                word_hidden.size(0)
            )
    
        # [B, max_words, D]
        padded_word_hidden = x.new_zeros(
            B,
            max_words,
            x.size(-1)
        )
    
        for b, word_hidden in enumerate(batch_word_hidden):
    
            n_words = word_hidden.size(0)
    
            padded_word_hidden[
                b, :n_words
            ] = word_hidden
    
        # [B, max_words, num_labels]
        logits = self.ner_head(padded_word_hidden)
    
        return logits
```

```python
def prepare_lner_data(text_data, gold_data, tokenizer):

    processed_data = {}

    for doc_id, text in text_data.items():

       

        # --------------------------------------------------
        # 1. Tokenize
        # --------------------------------------------------

        encoding = tokenizer.encode(text)

        tokens = encoding.tokens
        input_ids = encoding.ids
        offsets = encoding.offsets

        # --------------------------------------------------
        # 2. Group tokens
        #
        # Rule:
        #   Ġ  -> new word
        #   no Ġ -> continue previous word
        # --------------------------------------------------

        word_groups = []
        current_group = []

        for i, token in enumerate(tokens):

            start, end = offsets[i]

            if start == end:
                continue

            if current_group and token.startswith("Ġ"):
                word_groups.append(current_group)
                current_group = []

            current_group.append(i)

        if current_group:
            word_groups.append(current_group)

        # --------------------------------------------------
        # 3. Correct character spans
        #
        # Raw tokenizer offsets include the whitespace
        # represented by Ġ.
        # Remove that whitespace from the beginning.
        # --------------------------------------------------

        word_spans = []

        for group in word_groups:

            start = offsets[group[0]][0]
            end = offsets[group[-1]][1]

            while start < end and text[start].isspace():
                start += 1

            word_spans.append((start, end))

        # --------------------------------------------------
        # 4. Initialize labels
        # --------------------------------------------------

        word_labels = ["O"] * len(word_groups)

        # --------------------------------------------------
        # 5. Align gold entities
        # --------------------------------------------------
        if gold_data is not None:
            gold_entities = gold_data[doc_id]
            for entity_start, entity_end, entity_type, entity_text in gold_entities:
    
                overlapping_words = []
    
                for word_idx, (word_start, word_end) in enumerate(word_spans):
    
                    if word_start < entity_end and word_end > entity_start:
                        overlapping_words.append(word_idx)
    
                if not overlapping_words:
                    continue
    
                word_labels[overlapping_words[0]] = f"B-{entity_type}"
    
                for word_idx in overlapping_words[1:]:
                    word_labels[word_idx] = f"I-{entity_type}"

        gold_spans=[]
        if gold_data is not None:
            for a,b,label,entity_text in gold_data[doc_id]:
                if not (0<=a<b<=len(text)) or text[a:b]!=entity_text:
                    raise ValueError(f'Gold offset mismatch: {doc_id}, {(a,b)}')
                gold_spans.append([a,b,label])
        # --------------------------------------------------
        # 6. Save
        # --------------------------------------------------

        processed_data[doc_id] = {
            "input_ids": input_ids,
            "offsets": offsets,
            "word_groups": word_groups,
            "word_spans": word_spans,
            "word_labels": word_labels,
            "gold_spans": gold_spans,
            "text": text,
        }

    return processed_data
```

```python
MAX_SEQ_LEN = 512

def make_chunks(example, max_seq_len=512):
    """
    Split one processed document into chunks at word boundaries.

    Returns list of:
        {
            "input_ids": [...],
            "word_groups": [...],
            "word_labels": [...]
        }
    """

    input_ids = example["input_ids"]
    word_groups = example["word_groups"]
    word_labels = example["word_labels"]
    word_spans = example["word_spans"]
    chunks = []

    current_words = []
    current_start = None
    current_end = None

    for word_idx, group in enumerate(word_groups):

        word_start = group[0]
        word_end = group[-1] + 1

        if current_start is None:
            current_start = word_start
            current_end = word_end
            current_words = [word_idx]
            continue

        # Would adding this word exceed 2048?
        new_end = word_end

        if new_end - current_start > max_seq_len:

            # Save current chunk
            chunk_input_ids = input_ids[current_start:current_end]

            chunk_groups = [
                [
                    token_idx - current_start
                    for token_idx in word_groups[w]
                ]
                for w in current_words
            ]

            chunk_labels = [
                word_labels[w]
                for w in current_words
            ]
            chunk_spans = [
                word_spans[w]
                for w in current_words
            ]
            chunks.append({
                "input_ids": chunk_input_ids,
                "word_groups": chunk_groups,
                "word_labels": chunk_labels,
                "word_spans" : chunk_spans
            })

            # Start new chunk
            current_start = word_start
            current_end = word_end
            current_words = [word_idx]

        else:
            current_end = word_end
            current_words.append(word_idx)

    # Last chunk
    if current_words:

        chunk_input_ids = input_ids[current_start:current_end]

        chunk_groups = [
            [
                token_idx - current_start
                for token_idx in word_groups[w]
            ]
            for w in current_words
        ]

        chunk_labels = [
            word_labels[w]
            for w in current_words
        ]
        chunk_spans = [
            word_spans[w]
            for w in current_words
        ]
        chunks.append({
            "input_ids": chunk_input_ids,
            "word_groups": chunk_groups,
            "word_labels": chunk_labels,
            "word_spans" : chunk_spans
        })

    return chunks
```

```python
def collate_lner(batch, pad_token_id):

    max_len = max(
        len(x["input_ids"])
        for x in batch
    )

    max_words = max(
        len(x["word_labels"])
        for x in batch
    )

    input_ids = []
    word_groups = []
    word_labels = []

    for x in batch:

        # -------------------------
        # Token padding
        # -------------------------

        ids = x["input_ids"]

        padded = ids + [pad_token_id] * (
            max_len - len(ids)
        )

        input_ids.append(padded)

        # -------------------------
        # Word groups
        # -------------------------

        word_groups.append(
            x["word_groups"]
        )

        # -------------------------
        # Word-label padding
        # -------------------------

        labels = [
            label2id[label]
            for label in x["word_labels"]
        ]

        labels += [-100] * (
            max_words - len(labels)
        )

        word_labels.append(labels)

    return {
        "input_ids": torch.tensor(
            input_ids,
            dtype=torch.long
        ),

        "word_groups": word_groups,

        "word_labels": torch.tensor(
            word_labels,
            dtype=torch.long
        ),
    }
```

```python
def decode_bio_predictions(
    predictions,
    word_spans,
):
    entities = []

    current_label = None
    current_start = None
    current_end = None

    for label, (start, end) in zip(
        predictions,
        word_spans
    ):

        # O
        if label == "O":

            if current_label is not None:
                entities.append([
                    current_start,
                    current_end,
                    current_label
                ])

                current_label = None
                current_start = None
                current_end = None

            continue

        # B-ENTITY
        if label.startswith("B-"):

            if current_label is not None:
                entities.append([
                    current_start,
                    current_end,
                    current_label
                ])

            current_label = label[2:]
            current_start = start
            current_end = end

            continue

        # I-ENTITY
        if label.startswith("I-"):

            entity_type = label[2:]

            # Continue current entity
            if current_label == entity_type:

                current_end = end

            # Invalid I-ENTITY -> treat as a new entity
            else:

                if current_label is not None:
                    entities.append([
                        current_start,
                        current_end,
                        current_label
                    ])

                current_label = entity_type
                current_start = start
                current_end = end

    # Flush final entity
    if current_label is not None:
        entities.append([
            current_start,
            current_end,
            current_label
        ])

    return entities
```

```python
def bio_to_spans(labels, word_spans):
    """
    Convert BIO word labels + character spans into:
        [start, end, label]
    """

    entities = []

    current_label = None
    current_start = None
    current_end = None

    for label, (start, end) in zip(
        labels,
        word_spans
    ):

        if label == "O":

            if current_label is not None:
                entities.append([
                    current_start,
                    current_end,
                    current_label
                ])

                current_label = None
                current_start = None
                current_end = None

            continue

        if label.startswith("B-"):

            if current_label is not None:
                entities.append([
                    current_start,
                    current_end,
                    current_label
                ])

            current_label = label[2:]
            current_start = start
            current_end = end

        elif label.startswith("I-"):

            entity_type = label[2:]

            if current_label == entity_type:

                current_end = end

            else:
                # Treat invalid I-X as a new entity
                if current_label is not None:
                    entities.append([
                        current_start,
                        current_end,
                        current_label
                    ])

                current_label = entity_type
                current_start = start
                current_end = end

    if current_label is not None:
        entities.append([
            current_start,
            current_end,
            current_label
        ])

    return entities

from tqdm import tqdm
import torch


def generate_val_predictions(
    teacher,
    all_processed_data,
    val_doc_ids,
    id2label,
    max_seq_len=512,
):
    teacher.eval()

    predicted = {}
    gold = {}

    with torch.no_grad():

        for doc_id in tqdm(
            val_doc_ids,
            desc="Validation prediction"
        ):

            example = all_processed_data[doc_id]

            chunks = make_chunks(
                example,
                max_seq_len
            )

            doc_predictions = []

            for chunk in chunks:

                input_ids = torch.tensor(
                    [chunk["input_ids"]],
                    dtype=torch.long,
                    device=DEVICE
                )

                word_groups = [
                    chunk["word_groups"]
                ]

                logits = teacher(
                    input_ids,
                    word_groups
                )

                predicted_ids = logits.argmax(
                    dim=-1
                )[0]

                predicted_labels = [
                    id2label[idx.item()]
                    for idx in predicted_ids
                ]

                entities = decode_bio_predictions(
                    predicted_labels,
                    chunk["word_spans"]
                )

                doc_predictions.extend(
                    entities
                )

            predicted[doc_id] = doc_predictions

            gold[doc_id] = example['gold_spans']

    return predicted, gold

from collections import defaultdict


def evaluate_strict_ner(
    predicted,
    gold,
    labels,
):
    """
    Strict entity-level evaluation.

    Returns:
        Macro-F1
        Macro-Precision
        Macro-Recall
        per-label statistics
    """

    stats = {
        label: {
            "tp": 0,
            "fp": 0,
            "fn": 0,
        }
        for label in labels
    }

    for doc_id in gold:

        gold_entities = gold.get(
            doc_id,
            []
        )

        pred_entities = predicted.get(
            doc_id,
            []
        )

        # Exact entity tuples
        gold_set = set(
            tuple(entity)
            for entity in gold_entities
        )

        pred_set = set(
            tuple(entity)
            for entity in pred_entities
        )

        # True positives
        for entity in pred_set & gold_set:

            label = entity[2]

            if label in stats:
                stats[label]["tp"] += 1

        # False positives
        for entity in pred_set - gold_set:

            label = entity[2]

            if label in stats:
                stats[label]["fp"] += 1

        # False negatives
        for entity in gold_set - pred_set:

            label = entity[2]

            if label in stats:
                stats[label]["fn"] += 1

    # -------------------------
    # Per-label metrics
    # -------------------------

    per_label = {}

    macro_precision = 0.0
    macro_recall = 0.0
    macro_f1 = 0.0

    for label in labels:

        tp = stats[label]["tp"]
        fp = stats[label]["fp"]
        fn = stats[label]["fn"]

        precision = (
            tp / (tp + fp)
            if tp + fp > 0
            else 0.0
        )

        recall = (
            tp / (tp + fn)
            if tp + fn > 0
            else 0.0
        )

        f1 = (
            2 * precision * recall /
            (precision + recall)
            if precision + recall > 0
            else 0.0
        )

        per_label[label] = {
            "precision": precision,
            "recall": recall,
            "f1": f1,
            "tp": tp,
            "fp": fp,
            "fn": fn,
        }

        macro_precision += precision
        macro_recall += recall
        macro_f1 += f1

    n = len(labels)

    macro_precision /= n
    macro_recall /= n
    macro_f1 /= n

    return {
        "strict_macro_f1": macro_f1,
        "strict_macro_precision": macro_precision,
        "strict_macro_recall": macro_recall,
        "per_label": per_label,
    }
```

```python
def train_teacher(
    teacher,
    train_loader,
    val_loader,
    label2id,
    optimizer,
    trainable_params,
    tag,
    num_epochs=5,
    resume=True,
    resume_tag=None,
    patience=3,
    min_delta=0.001,
):
    device = next(teacher.parameters()).device

    criterion = nn.CrossEntropyLoss(
        ignore_index=-100
    )

    # =========================
    # RESUME
    # =========================

    if patience < 1 or min_delta < 0:
        raise ValueError("patience must be >= 1 and min_delta must be >= 0")
    start_epoch = 0
    best_f1 = float("-inf")
    patience_reference = float("-inf")
    epochs_without_improvement = 0
    history = []
    best_tag = f"{tag}_best"
    have_best = False

    if resume and resume_tag is not None:

        state = load_checkpoint(
            tag=resume_tag,
            model=teacher,
            optimizer=optimizer,
        )

        start_epoch = state["epoch"]
        extra = state.get("extra_state") or {}
        best_f1 = extra.get("best_f1", float("-inf"))
        patience_reference = extra.get("patience_reference", best_f1)
        epochs_without_improvement = extra.get("epochs_without_improvement", 0)
        history = extra.get("history", [])
        best_tag = extra.get("best_tag", best_tag)
        have_best = extra.get("have_best", False)
        # Older checkpoints have no early-stopping state: tracking starts here.

        print(
            f"Resuming from epoch {start_epoch}"
        )

    # =========================
    # TRAINING
    # =========================

    for epoch in range(start_epoch, num_epochs):
        if epochs_without_improvement >= patience:
            print("Resumed checkpoint already reached early-stopping patience")
            break

        teacher.train()

        total_train_loss = 0.0

        progress = tqdm(
            train_loader,
            desc=f"Epoch {epoch + 1}/{num_epochs}"
        )

        for step, batch in enumerate(progress):

            input_ids = batch["input_ids"].to(device)

            word_groups = batch["word_groups"]

            targets = batch["word_labels"].to(device)

            optimizer.zero_grad(set_to_none=True)

            # [B, max_words, num_labels]
            logits = teacher(
                input_ids,
                word_groups
            )

            # Flatten:
            # logits  -> [B * max_words, num_labels]
            # targets -> [B * max_words]
            loss = criterion(
                logits.view(
                    -1,
                    logits.size(-1)
                ),
                targets.view(-1)
            )

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                trainable_params,
                GRAD_CLIP
            )

            optimizer.step()

            total_train_loss += loss.item()

            progress.set_postfix(
                loss=f"{loss.item():.4f}"
            )

        avg_train_loss = (
            total_train_loss /
            len(train_loader)
        )

        # =========================
        # VALIDATION
        # =========================

        teacher.eval()

        total_val_loss = 0.0
        
        with torch.no_grad():
        
            for batch in val_loader:
        
                input_ids = batch["input_ids"].to(device)
                word_groups = batch["word_groups"]
                targets = batch["word_labels"].to(device)
        
                logits = teacher(
                    input_ids,
                    word_groups
                )
        
                loss = criterion(
                    logits.view(-1, logits.size(-1)),
                    targets.view(-1)
                )
        
                total_val_loss += loss.item()
        
        avg_val_loss = (
            total_val_loss /
            len(val_loader)
        )
        
        # =========================
        # STRICT VALIDATION F1
        # =========================
        
        val_predicted, val_gold = generate_val_predictions(
            teacher=teacher,
            all_processed_data=all_processed_data,
            val_doc_ids=val_doc_ids,
            id2label=id2label,
            max_seq_len=MAX_SEQ_LEN,
        )
        
        val_metrics = evaluate_strict_ner(
            predicted=val_predicted,
            gold=val_gold,
            labels=ENTITY_LABELS,
        )
        
        val_f1 = val_metrics["strict_macro_f1"]
        # =========================
        # LOGGING
        # =========================

        print(
            f"\nEpoch {epoch + 1}/{num_epochs}"
        )

        print(
            f"Train Loss:    {avg_train_loss:.4f}"
        )

        print(
            f"Val Loss:      {avg_val_loss:.4f}"
        )

        print(
            f"Val Strict F1:  {val_f1:.4f}"
        )

        # Save any strict F1 improvement, even if it is smaller than min_delta.
        improved = val_f1 > best_f1
        if improved:
            best_f1 = val_f1
            have_best = True

        # min_delta controls patience resets, not which checkpoint is best.
        if val_f1 > patience_reference + min_delta:
            patience_reference = val_f1
            epochs_without_improvement = 0
        else:
            epochs_without_improvement += 1

        history.append({
            "epoch": epoch + 1,
            "train_loss": avg_train_loss,
            "val_loss": avg_val_loss,
            "val_strict_macro_f1": val_f1,
        })
        extra = {
            "best_f1": best_f1,
            "patience_reference": patience_reference,
            "epochs_without_improvement": epochs_without_improvement,
            "best_tag": best_tag,
            "have_best": have_best,
            "history": history,
        }
        if improved:
            save_checkpoint(
                tag=best_tag, model=teacher, optimizer=optimizer,
                epoch=epoch + 1, extra_state=extra,
            )
            print(f"New best checkpoint: {best_tag} (F1={best_f1:.4f})")

        # Keep latest state separately so an interrupted run can resume.
        save_checkpoint(
            tag=tag, model=teacher, optimizer=optimizer,
            epoch=epoch + 1, extra_state=extra,
        )
        print(f"Early stopping: {epochs_without_improvement}/{patience}")
        if epochs_without_improvement >= patience:
            print(f"Stopping early. Best validation F1: {best_f1:.4f}")
            break

    if have_best:
        # Mutates the existing teacher, so the existing prediction cells use best weights.
        load_checkpoint(tag=best_tag, model=teacher, optimizer=optimizer)
        print(f"Restored best model from {best_tag} (F1={best_f1:.4f})")
    teacher.eval()
    return teacher, history
```

```python
import json
import zipfile
import os
import torch
from tqdm import tqdm


def generate_predictions(
    teacher,
    processed_test,
    id2label,
    tag,
    max_seq_len=512,
):
    teacher.eval()

    predictions = {}

    with torch.no_grad():

        for doc_id, example in tqdm(
            processed_test.items(),
            desc="Generating predictions"
        ):

            chunks = make_chunks(
                example,
                max_seq_len
            )

            document_entities = []

            for chunk in chunks:

                input_ids = torch.tensor(
                    [chunk["input_ids"]],
                    dtype=torch.long,
                    device=DEVICE
                )

                word_groups = [
                    chunk["word_groups"]
                ]

                logits = teacher(
                    input_ids,
                    word_groups
                )

                predicted_ids = logits.argmax(
                    dim=-1
                )[0]

                predicted_labels = [
                    id2label[idx.item()]
                    for idx in predicted_ids
                ]

                entities = decode_bio_predictions(
                    predicted_labels,
                    chunk["word_spans"]
                )

                document_entities.extend(
                    entities
                )

            predictions[doc_id] = document_entities

    # =========================
    # Save predictions.json
    # =========================

    json_path = f"/kaggle/working/predictions.json"

    with open(json_path, "w") as f:
        json.dump(
            predictions,
            f,
            indent=2
        )

    # =========================
    # Zip as tag.zip
    # =========================

    zip_path = f"/kaggle/working/{tag}.zip"

    with zipfile.ZipFile(
        zip_path,
        "w",
        zipfile.ZIP_DEFLATED
    ) as z:

        z.write(
            json_path,
            arcname="prediction.json"
        )

    print(f"Saved: {zip_path}")

    return zip_path
```
