```python
def upload_to_bucket(local_path, remote_path):
    remote_path = remote_path.lstrip("/")

    fs.put_file(
        local_path,
        f"{BUCKET_ROOT}/{remote_path}",
    )

    print(f"Uploaded: {remote_path}")
```

```python
def download_from_bucket(remote_path, local_path):
    remote_path = remote_path.lstrip("/")

    fs.get_file(
        f"{BUCKET_ROOT}/{remote_path}",
        local_path,
    )

    print(f"Downloaded: {remote_path} -> {local_path}")
```

```python
def upload_checkpoint(tmp_dir, tag):

    remote_dir = f"checkpoints/{tag}"

    for filename in os.listdir(tmp_dir):
        local_path = os.path.join(tmp_dir, filename)

        upload_to_bucket(
            local_path,
            f"{remote_dir}/{filename}",
        )
```

```python
def save_checkpoint(tag, model, optimizer, scheduler=None, scaler=None,
                    tokens_seen=0, global_step=0, epoch=0, extra_state=None):

    with tempfile.TemporaryDirectory() as tmp_dir:

        torch.save(model.state_dict(), os.path.join(tmp_dir, "model.pt"))
        torch.save(optimizer.state_dict(), os.path.join(tmp_dir, "optimizer.pt"))

        if scheduler is not None:
            torch.save(scheduler.state_dict(), os.path.join(tmp_dir, "scheduler.pt"))

        if scaler is not None:
            torch.save(scaler.state_dict(), os.path.join(tmp_dir, "scaler.pt"))

        rng_state = {"python": random.getstate(), "torch": torch.get_rng_state()}

        if torch.cuda.is_available():
            rng_state["cuda"] = torch.cuda.get_rng_state_all()

        torch.save(rng_state, os.path.join(tmp_dir, "rng_state.pt"))

        state = {
            "tag": tag,
            "tokens_seen": tokens_seen,
            "global_step": global_step,
            "epoch": epoch,
            "extra_state": extra_state,
        }

        with open(os.path.join(tmp_dir, "training_state.json"), "w") as f:
            json.dump(state, f, indent=2)

        upload_checkpoint(tmp_dir,tag)

def load_checkpoint(tag, model, optimizer, scheduler=None, scaler=None):

    with tempfile.TemporaryDirectory() as tmp_dir:

        files = [
            "model.pt",
            "optimizer.pt",
            "scheduler.pt",
            "scaler.pt",
            "rng_state.pt",
            "training_state.json",
        ]

        checkpoint_dir = os.path.join(
            tmp_dir,
            "checkpoints",
            tag,
        )

        for filename in files:
            try:
                download_from_bucket(
                    f"checkpoints/{tag}/{filename}",
                    os.path.join(checkpoint_dir, filename),
                )
            except Exception:
                pass

        model.load_state_dict(
            torch.load(
                os.path.join(checkpoint_dir, "model.pt"),
                map_location=DEVICE,
                weights_only=True,
            )
        )

        optimizer.load_state_dict(
            torch.load(
                os.path.join(checkpoint_dir, "optimizer.pt"),
                map_location=DEVICE,
                weights_only=False,
            )
        )

        scheduler_path = os.path.join(checkpoint_dir, "scheduler.pt")

        if scheduler is not None and os.path.exists(scheduler_path):
            scheduler.load_state_dict(
                torch.load(
                    scheduler_path,
                    map_location=DEVICE,
                    weights_only=False,
                )
            )

        scaler_path = os.path.join(checkpoint_dir, "scaler.pt")

        if scaler is not None and os.path.exists(scaler_path):
            scaler.load_state_dict(
                torch.load(
                    scaler_path,
                    map_location=DEVICE,
                    weights_only=False,
                )
            )

        rng_state = torch.load(
            os.path.join(checkpoint_dir, "rng_state.pt"),
            map_location="cpu",
            weights_only=False,
        )

        random.setstate(rng_state["python"])
        torch.set_rng_state(rng_state["torch"])

        if torch.cuda.is_available() and "cuda" in rng_state:
            torch.cuda.set_rng_state_all(rng_state["cuda"])

        with open(
            os.path.join(checkpoint_dir, "training_state.json")
        ) as f:
            state = json.load(f)

    return state
```

```python
def add_custom_legal_tokens(tokenizer, corpus_path, min_frequency=100):
    patterns = [
        re.compile(r"\b(?:Article|ARTICLE)\s+\d+[A-Za-z]*(?:\([0-9A-Za-z]+\))*"),
        re.compile(r"\b(?:Section|SECTION)\s+\d+[A-Za-z]*(?:\([0-9A-Za-z]+\))*"),
        re.compile(r"\b(?:Rule|RULE)\s+\d+[A-Za-z]*(?:\([0-9A-Za-z]+\))*"),
        re.compile(r"\b(?:Regulation|REGULATION)\s+\d+[A-Za-z]*(?:\([0-9A-Za-z]+\))*"),
        re.compile(r"\([a-z]\)"),
        re.compile(r"\([0-9]{1,2}\)"),
        re.compile(r"\((?:i{1,3}|iv|v|vi{0,3}|ix|x{1,2})\)"),
    ]

    counts = Counter()

    for text in corpus_iterator(corpus_path):
        for pattern in patterns:
            counts.update(pattern.findall(text))

    custom_tokens = []

    for token, count in counts.items():
        token_with_space = "Ġ" + token

        if count >= min_frequency and tokenizer.token_to_id(token_with_space) is None:
            custom_tokens.append(token_with_space)

    tokenizer.add_tokens(custom_tokens)

    return custom_tokens, counts
```

## Model - A1 

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


# ============================================================
# RoPE
# ============================================================

class RotaryEmbedding(nn.Module):
    def __init__(self, head_dim, max_seq_len=2048, base=10000):
        super().__init__()

        inv_freq = 1.0 / (
            base ** (
                torch.arange(0, head_dim, 2).float() / head_dim
            )
        )

        positions = torch.arange(max_seq_len).float()
        freqs = torch.outer(positions, inv_freq)

        self.register_buffer(
            "cos",
            freqs.cos(),
            persistent=False,
        )

        self.register_buffer(
            "sin",
            freqs.sin(),
            persistent=False,
        )

    def forward(self, q, k):

        T = q.size(2)

        cos = self.cos[:T].unsqueeze(0).unsqueeze(0)
        sin = self.sin[:T].unsqueeze(0).unsqueeze(0)

        q1, q2 = q[..., ::2], q[..., 1::2]
        k1, k2 = k[..., ::2], k[..., 1::2]

        q_rot = torch.stack(
            [
                q1 * cos - q2 * sin,
                q1 * sin + q2 * cos,
            ],
            dim=-1,
        ).flatten(-2)

        k_rot = torch.stack(
            [
                k1 * cos - k2 * sin,
                k1 * sin + k2 * cos,
            ],
            dim=-1,
        ).flatten(-2)

        return q_rot, k_rot


# ============================================================
# SwiGLU Expert
# ============================================================

class SwiGLU(nn.Module):

    def __init__(self, d_model, d_ff):
        super().__init__()

        self.gate_proj = nn.Linear(
            d_model,
            d_ff,
            bias=False,
        )

        self.up_proj = nn.Linear(
            d_model,
            d_ff,
            bias=False,
        )

        self.down_proj = nn.Linear(
            d_ff,
            d_model,
            bias=False,
        )

    def forward(self, x):

        return self.down_proj(
            F.silu(self.gate_proj(x))
            * self.up_proj(x)
        )


# ============================================================
# Mixture of Experts
# ============================================================
class MoE(nn.Module):

    def __init__(
        self,
        d_model,
        d_ff,
        n_experts=8,
        top_k=2,
        balance_alpha=0.003,
    ):
        super().__init__()

        self.n_experts = n_experts
        self.top_k = top_k
        self.balance_alpha = balance_alpha

        # Router chooses experts BEFORE any expert is evaluated
        self.router = nn.Linear(
            d_model,
            n_experts,
            bias=False,
        )

        expert_d_ff = d_ff // top_k

        self.experts = nn.ModuleList([
            SwiGLU(
                d_model,
                expert_d_ff,
            )
            for _ in range(n_experts)
        ])

    def forward(self, x):

        # x: [B, T, D]

        # ------------------------------------------------
        # Batch representation
        # ------------------------------------------------

        batch_repr = x.mean(
            dim=(0, 1)
        )  # [D]

        # ------------------------------------------------
        # ROUTE FIRST
        # ------------------------------------------------

        router_logits = self.router(
            batch_repr
        )  # [E]

        router_probs = F.softmax(
            router_logits,
            dim=-1,
        )

        topk_probs, topk_indices = torch.topk(
            router_probs,
            self.top_k,
            dim=-1,
        )

        # Normalize selected weights
        topk_probs = topk_probs / (
            topk_probs.sum() + 1e-9
        )

        # ------------------------------------------------
        # ONLY NOW evaluate selected experts
        # ------------------------------------------------

        output = 0.0

        for i in range(self.top_k):

            expert_idx = topk_indices[i].item()
            weight = topk_probs[i]

            expert_output = self.experts[
                expert_idx
            ](x)

            output = (
                output
                + weight * expert_output
            )

        # ------------------------------------------------
        # Batch routing loss
        # ------------------------------------------------

        # Router probability distribution for this batch.
        # Used to discourage permanent collapse to
        # a small subset of experts.
        balance_loss = (
            self.balance_alpha
            * self.n_experts
            * torch.sum(
                router_probs
                * router_probs
            )
        )

        return output, balance_loss

# ============================================================
# GQA + SDPA
# ============================================================
class GQAAttention(nn.Module):

    def __init__(
        self,
        d_model,
        n_heads,
        n_kv_heads,
        local_window=64,
        global_stride=128,
        max_seq_len=2048,
    ):
        super().__init__()

        assert d_model % n_heads == 0
        assert n_heads % n_kv_heads == 0

        self.n_heads = n_heads
        self.n_kv_heads = n_kv_heads
        self.local_window = local_window
        self.global_stride = global_stride
        self.max_seq_len = max_seq_len

        self.head_dim = d_model // n_heads

        self.q_proj = nn.Linear(
            d_model,
            n_heads * self.head_dim,
            bias=False,
        )

        self.k_proj = nn.Linear(
            d_model,
            n_kv_heads * self.head_dim,
            bias=False,
        )

        self.v_proj = nn.Linear(
            d_model,
            n_kv_heads * self.head_dim,
            bias=False,
        )

        self.out_proj = nn.Linear(
            d_model,
            d_model,
            bias=False,
        )

        self.rope = RotaryEmbedding(
            self.head_dim,
            max_seq_len=max_seq_len,
        )

    def _build_sparse_mask(self, T, device):
        """
        Allowed attention for query position i:

        1. Previous local_window positions
        2. All previous global positions
           (every global_stride tokens)
        3. Itself

        Future positions are never allowed.
        """

        positions = torch.arange(
            T,
            device=device,
        )

        # i = query position
        # j = key position
        i = positions[:, None]
        j = positions[None, :]

        # Causal condition
        causal = j <= i

        # Local attention
        local = j >= (i - self.local_window + 1)

        # Global landmark positions
        global_pos = (j % self.global_stride) == 0

        # Allowed if local OR global, while remaining causal
        allowed = causal & (local | global_pos)

        # SDPA bool mask:
        # True  = allowed to attend
        # False = masked
        return allowed

    def forward(self, x):

        B, T, _ = x.shape

        # --------------------------------------------------
        # Q, K, V projections
        # --------------------------------------------------

        q = self.q_proj(x)
        k = self.k_proj(x)
        v = self.v_proj(x)

        # --------------------------------------------------
        # Reshape into heads
        # --------------------------------------------------

        q = q.view(
            B,
            T,
            self.n_heads,
            self.head_dim,
        ).transpose(1, 2)

        k = k.view(
            B,
            T,
            self.n_kv_heads,
            self.head_dim,
        ).transpose(1, 2)

        v = v.view(
            B,
            T,
            self.n_kv_heads,
            self.head_dim,
        ).transpose(1, 2)

        # --------------------------------------------------
        # RoPE
        # --------------------------------------------------

        q, k = self.rope(q, k)

        # --------------------------------------------------
        # GQA: expand K/V heads
        # --------------------------------------------------

        repeat_factor = (
            self.n_heads // self.n_kv_heads
        )

        k = k.repeat_interleave(
            repeat_factor,
            dim=1,
        )

        v = v.repeat_interleave(
            repeat_factor,
            dim=1,
        )

        # --------------------------------------------------
        # Local + Global sparse attention
        # --------------------------------------------------

        attn_mask = self._build_sparse_mask(
            T,
            x.device,
        )

        attn_out = F.scaled_dot_product_attention(
            q,
            k,
            v,
            attn_mask=attn_mask,
            is_causal=False,
        )

        # --------------------------------------------------
        # Merge heads
        # --------------------------------------------------

        attn_out = (
            attn_out
            .transpose(1, 2)
            .contiguous()
            .view(B, T, -1)
        )

        return self.out_proj(attn_out)


# ============================================================
# Decoder Block with MoE
# ============================================================

class DecoderBlock(nn.Module):

    def __init__(
        self,
        d_model,
        n_heads,
        n_kv_heads,
        d_ff,
        n_experts=8,
        top_k=2,
        balance_alpha=0.003,
    ):
        super().__init__()

        self.norm1 = nn.RMSNorm(
            d_model
        )

        self.attention = GQAAttention(
            d_model,
            n_heads,
            n_kv_heads,
        )

        self.norm2 = nn.RMSNorm(
            d_model
        )

        self.moe = MoE(
            d_model=d_model,
            d_ff=d_ff,
            n_experts=n_experts,
            top_k=top_k,
            balance_alpha=balance_alpha,
        )

    def forward(self, x):

        # Attention
        x = x + self.attention(
            self.norm1(x)
        )

        # MoE FFN
        moe_output, balance_loss = self.moe(
            self.norm2(x)
        )

        x = x + moe_output

        return x, balance_loss


# ============================================================
# Decoder-only MoE Language Model
# ============================================================

class EfficientDecoderLM(nn.Module):

    def __init__(
        self,
        vocab_size,
        d_model=768,
        n_heads=12,
        n_kv_heads=4,
        n_layers=12,
        d_ff=3072,
        max_seq_len=2048,

        # MoE
        n_experts=8,
        top_k=2,
        balance_alpha=0.003,
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.d_model = d_model

        self.n_experts = n_experts
        self.top_k = top_k
        self.balance_alpha = balance_alpha

        self.token_embedding = nn.Embedding(
            vocab_size,
            d_model,
        )

        self.blocks = nn.ModuleList([
            DecoderBlock(
                d_model=d_model,
                n_heads=n_heads,
                n_kv_heads=n_kv_heads,
                d_ff=d_ff,
                n_experts=n_experts,
                top_k=top_k,
                balance_alpha=balance_alpha,
            )
            for _ in range(n_layers)
        ])

        self.final_norm = nn.RMSNorm(
            d_model
        )

        # Tied LM head
        self.lm_head = nn.Linear(
            d_model,
            vocab_size,
            bias=False,
        )

        self.lm_head.weight = (
            self.token_embedding.weight
        )

        self.apply(
            self._init_weights
        )

    def _init_weights(self, module):

        if isinstance(module, nn.Linear):

            nn.init.xavier_uniform_(
                module.weight
            )

        elif isinstance(module, nn.Embedding):

            nn.init.xavier_uniform_(
                module.weight
            )

        elif isinstance(module, nn.RMSNorm):

            nn.init.ones_(
                module.weight
            )

    def forward(
        self,
        input_ids,
        return_aux_loss=False,
    ):

        x = self.token_embedding(
            input_ids
        )

        total_balance_loss = x.new_zeros(
            ()
        )

        for block in self.blocks:

            x, balance_loss = block(x)

            total_balance_loss = (
                total_balance_loss
                + balance_loss
            )

        x = self.final_norm(x)

        logits = self.lm_head(x)

        if return_aux_loss:

            return (
                logits,
                total_balance_loss,
            )

        return logits
```

```python
import time
import torch
import torch.nn.functional as F
from bert_score import score as bert_score


def train(
    model,
    train_loader,
    optimizer,
    val_data,
    train_tokens,
    tag,
    val_interval_tokens=1_000_000,
    resume=False,
    resume_tag=None,
):
    model.train()

    tokens_seen = 0
    global_step = 0
    epoch = 0
    next_val_tokens = val_interval_tokens

    if resume:
        if resume_tag is None:
            raise ValueError("resume_tag must be provided when resume=True")

        state = load_checkpoint(
            tag=resume_tag,
            model=model,
            optimizer=optimizer,
        )

        tokens_seen = state["tokens_seen"]
        global_step = state["global_step"]
        epoch = state["epoch"]

        next_val_tokens = (
            (tokens_seen // val_interval_tokens) + 1
        ) * val_interval_tokens

        print(
            f"Resumed from {resume_tag} | "
            f"Tokens: {tokens_seen:,} | "
            f"Step: {global_step}"
        )

    start_time = time.perf_counter()

    while tokens_seen < train_tokens:

        for input_ids in train_loader:

            if tokens_seen >= train_tokens:
                break

            input_ids = input_ids.to(DEVICE, non_blocking=True)

            inputs = input_ids[:, :-1]
            targets = input_ids[:, 1:]
        
            optimizer.zero_grad(
                set_to_none=True
            )
        
            logits, balance_loss = model(
                inputs,
                return_aux_loss=True,
            )
        
            lm_loss = F.cross_entropy(
                logits.reshape(
                    -1,
                    model.vocab_size,
                ),
                targets.reshape(-1),
            )
        
            loss = lm_loss + balance_loss
            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                1.0,
            )

            optimizer.step()

            batch_tokens = input_ids.numel()
            tokens_seen += batch_tokens
            global_step += 1

            torch.cuda.synchronize()

            elapsed = time.perf_counter() - start_time
            tokens_per_sec = tokens_seen / elapsed

            print(
                f"Step {global_step:5d} | "
                f"Tokens {tokens_seen:,}/{train_tokens:,} | "
                f"Loss {loss.item():.4f} | "
                f"Time {elapsed:.2f}s | "
                f"Tokens/s {tokens_per_sec:,.0f}"
            )
            
            # inside train(), replace the generation/BERTScore validation block with:
            
            if tokens_seen >= next_val_tokens:
            
                model.eval()
            
                val_loss = 0.0
                val_tokens = 0
            
                with torch.no_grad():
                    for example in val_data:
            
                        prompt_ids = tokenizer.encode(example["prompt"]).ids
                        target_ids = tokenizer.encode(example["target"]).ids
            
                        input_ids = torch.tensor(
                            [prompt_ids + target_ids],
                            dtype=torch.long,
                            device=DEVICE,
                        )
            
                        logits = model(input_ids,return_aux_loss=False)
            
                        prompt_len = len(prompt_ids)
            
                        target_logits = logits[:, prompt_len - 1:-1, :]
                        target = input_ids[:, prompt_len:]
            
                        loss = F.cross_entropy(
                            target_logits.reshape(-1, target_logits.size(-1)),
                            target.reshape(-1),
                            reduction="sum",
                        )
            
                        val_loss += loss.item()
                        val_tokens += target.numel()
            
                val_loss /= val_tokens
            
                print(
                    f"Validation @ {tokens_seen:,} tokens | "
                    f"Val Loss: {val_loss:.4f}"
                )
            
                model.train()
            
                save_checkpoint(
                    tag=tag,
                    model=model,
                    optimizer=optimizer,
                    tokens_seen=tokens_seen,
                    global_step=global_step,
                    epoch=epoch,
                )
            
                print(f"Checkpoint saved: {tag}\n")
            
                next_val_tokens += val_interval_tokens

    total_time = time.perf_counter() - start_time

    print("\nTraining complete.")
    print(f"Tokens seen: {tokens_seen:,}")
    print(f"Total time: {total_time:.2f}s")
    print(f"Average tokens/s: {tokens_seen / total_time:,.0f}")
```

```python
@torch.no_grad()
def generate_single(
    prompt,
    model,
    tokenizer,
    device,
    target_words=128,
    temperature=0.8,
    top_p=0.9,
):
    model.eval()

    max_input_len = SEQ_LEN - MAX_GENERATION_TOKENS

    prompt_ids = tokenizer.encode(prompt).ids[:max_input_len]

    input_ids = torch.tensor(
        [prompt_ids],
        dtype=torch.long,
        device=device,
    )

    generated_ids = []

    # ========================================================
    # Autoregressive generation
    # ========================================================

    for _ in range(MAX_GENERATION_TOKENS):

        # ----------------------------------------------------
        # Run the ENTIRE sequence every time
        # ----------------------------------------------------

        logits = model(input_ids,return_aux_loss = False)

        next_logits = (
            logits[:, -1, :]
            / temperature
        )

        # ----------------------------------------------------
        # Top-p sampling
        # ----------------------------------------------------

        sorted_logits, sorted_indices = torch.sort(
            next_logits,
            descending=True,
            dim=-1,
        )

        sorted_probs = torch.softmax(
            sorted_logits,
            dim=-1,
        )

        cumulative_probs = torch.cumsum(
            sorted_probs,
            dim=-1,
        )

        remove = cumulative_probs > top_p

        remove[:, 1:] = remove[:, :-1].clone()
        remove[:, 0] = False

        sorted_logits[remove] = float("-inf")

        probs = torch.softmax(
            sorted_logits,
            dim=-1,
        )

        sampled = torch.multinomial(
            probs,
            1,
        )

        next_token = torch.gather(
            sorted_indices,
            -1,
            sampled,
        )

        # ----------------------------------------------------
        # Store generated token
        # ----------------------------------------------------

        generated_ids.append(
            next_token.item()
        )

        # ----------------------------------------------------
        # Stop at 128 words
        # ----------------------------------------------------

        text = tokenizer.decode(generated_ids)

        if len(text.split()) >= target_words:
            break

        # ----------------------------------------------------
        # Append token to the FULL sequence
        # ----------------------------------------------------

        input_ids = torch.cat(
            [
                input_ids,
                next_token,
            ],
            dim=1,
        )

    # ========================================================
    # Final prediction
    # ========================================================

    text = tokenizer.decode(generated_ids)

    text = " ".join(
        text.split()[:target_words]
    )

    return text
```
